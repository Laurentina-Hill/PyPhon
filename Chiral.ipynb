## Chiral Phonons

### Diatomic (Honeycomb) Lattice

Fig 1: Display of a diatomic honeycomb lattice. Blue points represent atom A in these calculations. Specifically the right most atom A in each hexagonal array

#### Description of the Honeycomb Lattice
The honeycomb lattice has two standard basis vectors stemming from the left most atom in each hexagon. These vectors are denoted as
$$
\vec{a_1} = a \begin{pmatrix} \frac{3}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix} \quad 
\vec{a_2} = a \begin{pmatrix} \frac{3}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix}
$$

It should be noted that the value $a$ for unit cell spacing is not the same as the vectors  
The honeycomb lattice has bond displacements similar to that of a triangle lattice (the system can be seen as an overlap of 2 triangular lattice, but that is not covered here). Describing the displacements in terms of the basis vectors can be done as such    
The $1st$ nearest-neighbors vectors are all $A$-$B$ bonds between alternating atom types
$$
\begin{align}
\vec{h_1} = a \begin{pmatrix} 1 \\ 0 \end{pmatrix} \quad
\vec{h_2} = \vec{h_1} - \vec{a_1} = a \begin{pmatrix} -\frac{1}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix} \quad
\vec{h_3} = \vec{h_1} - \vec{a_2} = a \begin{pmatrix} -\frac{1}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix} \quad
\end{align}
$$
All of which simply have directions 
$$
\hat{h_j} = \frac{\vec{h_j}}{a}
$$
 
The $2nd$ nearest-neighbor are all like atom $A$-$A$ bonds. There are $6$ of these bonds and they are the same as a standard triangular lattice
$$
\begin{align}
\vec{p}_{1,3} = \pm \vec{a_1} = \pm a \begin{pmatrix} \frac{3}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix} \quad
\vec{p}_{2,3} = \pm \vec{a_2} = \pm a \begin{pmatrix} \frac{3}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix} \quad
\vec{p}_{5,6} = \pm (\vec{a_1} - \vec{a_2}) = \pm a \begin{pmatrix} 0 \\ \sqrt{3} \end{pmatrix} 
\end{align}
$$
With the corresponding directions
$$
\begin{align}
\hat{p}_{1,3} = \pm \begin{pmatrix} \frac{\sqrt{3}}{2} \\ \frac{1}{2} \end{pmatrix} \quad
\hat{p}_{2,3} = \pm \begin{pmatrix} \frac{\sqrt{3}}{2} \\ -\frac{1}{2} \end{pmatrix} \quad
\hat{p}_{5,6} = \pm \begin{pmatrix} 0 \\ 1 \end{pmatrix} 
\end{align}
$$

Finally, there are 3 $3rd$ nearest-neighbors which exist in opposite positions to the $1st$ nearest-neighbors as more $A$-$B$ bonds, however these have lengths of $2a$ now and reach across the hexagonal plane in opposite directions from $\vec{h_j}$
$$
\begin{align}
\vec{b_1} = -2\hat{h_1} \quad
\vec{b_2} = \vec{h_2} + \vec{a_2} = a \begin{pmatrix} 1 \\ -\sqrt{3} \end{pmatrix} \quad
\vec{b_3} = \vec{h_3} + \vec{a_1} = a \begin{pmatrix} 1 \\ \sqrt{3} \end{pmatrix} \quad
\end{align}
$$
With the corresponding directions
$$
\begin{align}
\hat{b_1} = \begin{pmatrix} -1 \\ 0 \end{pmatrix} \quad
\hat{b_2} = \begin{pmatrix} \frac{1}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix} \quad
\hat{b_3} = \begin{pmatrix} \frac{1}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix}
\end{align}
$$

#### Hooke's Law with Projection 
A spring connecting $R$ to $R + \delta$ represents a change in length along the $\hat{e}$ direction
$$
\begin{equation} \Delta \ell (\vec{R},\vec{\delta}) = \hat{e} \cdot [\vec{U}(\vec{R} + \vec{\delta}) - \vec{U}] \end{equation}
\tag{1}
$$

where the force causing this displacement is represented by some spring constant, $C$, along a direction, $\hat{e}$
$$
\begin{equation} \vec{F} (\vec{R},\vec{\delta}) = C \Delta \ell (\vec{R},\vec{\delta})\hat{e} = C [\hat{e} \cdot [\vec{U}(\vec{R} + \vec{\delta}) - \vec{U}]]\hat{e} \end{equation}
\tag{2}
$$

Using the vector relationship that
$$
\begin{equation} \vec{A}(\vec{A} \cdot \vec{B}) = (\vec{A} \vec{A}^T)\vec{B} = (\vec{A} \otimes \vec{A})\vec{B} \end{equation}
$$

the equation can be rearrange in terms of a projection on the $\hat{e}$ direction. Note that multiplying a vector by it's transpose is the same as the outer product notation. The force is now 
$$
\begin{equation} \vec{F} (\vec{R},\vec{\delta}) = C(\hat{e} \otimes \hat{e})[\vec{U}(\vec{R} + \vec{\delta}) - \vec{U}] \end{equation}
\tag{3}
$$

Now using Newton's $2nd$ Law the force on atom $A$ due to a displacement of atom $B$ is represented by 
$$
\begin{equation} 
M \ddot{U}_A(\vec{R}) = 
C_1 \sum_{j=1}^{3} (\hat{h_j} \otimes \hat{h_j}) [\vec{U_B}(\vec{R} + \vec{h_j}) - \vec{U_A}] +
C_3 \sum_{j=1}^{3} (\hat{b_j} \otimes \hat{b_j}) [\vec{U_B}(\vec{R} + \vec{b_j}) - \vec{U_A}]
\end{equation}
\tag{4}
$$

And the forces on like atoms $A$-$A$ is seen as 
$$
\begin{equation} 
M \ddot{U}_A(\vec{R}) = 
C_2 \sum_{j=1}^{6} (\hat{p_j} \otimes \hat{p_j}) [\vec{U_A}(\vec{R} + \vec{p_j}) - \vec{U_A}]
\end{equation}
\tag{5}
$$

#### Projectors
Before making trying for a solution to the system the projectors should be knows to simplify calculations
$$
\begin{align}
&(\hat{h_1} \otimes \hat{h_1}) = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} \quad
(\hat{h_2} \otimes \hat{h_2}) = \begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} \quad
(\hat{h_3} \otimes \hat{h_3}) = \begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} \\ 
\\
&(\hat{p}_{1,3} \otimes \hat{p}_{1,3}) = \begin{pmatrix} \frac{3}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} \quad
(\hat{p}_{2,4} \otimes \hat{p}_{2,4}) = \begin{pmatrix} \frac{3}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} \quad
(\hat{p}_{5,6} \otimes \hat{p}_{5,6}) = \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} \\
\\
&(\hat{b_1} \otimes \hat{b_1}) = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} \quad
(\hat{b_2} \otimes \hat{b_2}) = \begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} \quad
(\hat{b_3} \otimes \hat{b_3}) = \begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} 
\end{align}
$$

#### Bloch's Theorem 
Applying Bloch's theorem says that any wavefunction in a periodic crystal can be described as a moving wave of the form 
$$
\vec{U}(\vec{R},t) = \vec{U}e^{i(\vec{k} \cdot \vec{R} -\omega t)}
$$

Applying this to the atoms in this system yields
$$
\vec{U_A}(\vec{R},t) = \vec{U_A}e^{i(\vec{k} \cdot \vec{R} -\omega t)} \quad
\vec{U_B}(\vec{R},t) = \vec{U_B}e^{i(\vec{k} \cdot \vec{R} -\omega t)} \quad
\vec{U_B}(\vec{R} + \vec{\delta},t) = \vec{U_B}(\vec{R},t) e^{i(\vec{k} \cdot \vec{\delta})} 
$$

Which when plugged into $Eq4$ and $Eq5$ yield
$$
\begin{align} 
-M \omega^2 \vec{U_A} = 
C_1 \sum_{j=1}^{3} (\hat{h_j} \otimes \hat{h_j}) [\vec{U_B} e^{i(\vec{k} \cdot \vec{h_j})} - \vec{U_A}] +
C_3 \sum_{j=1}^{3} (\hat{b_j} \otimes \hat{b_j}) [\vec{U_B} e^{i(\vec{k} \cdot \vec{b_j})} - \vec{U_A}] \\
M \omega^2 \vec{U_A} = 
C_1 \sum_{j=1}^{3} (\hat{h_j} \otimes \hat{h_j}) [\vec{U_A} - \vec{U_B} e^{i(\vec{k} \cdot \vec{h_j})}] +
C_3 \sum_{j=1}^{3} (\hat{b_j} \otimes \hat{b_j}) [\vec{U_A} - \vec{U_B} e^{i(\vec{k} \cdot \vec{b_j})}]
\end{align}
\tag{6}
$$
$$
\begin{align} 
-M \omega^2 \vec{U_A} = 
C_2 \sum_{j=1}^{6} (\hat{p_j} \otimes \hat{p_j}) [\vec{U_A}e^{i(\vec{k} \cdot \vec{p_j})} - \vec{U_A}] \\
M \omega^2 \vec{U_A} = 
C_2 \sum_{j=1}^{6} (\hat{p_j} \otimes \hat{p_j}) [\vec{U_A} - \vec{U_A}e^{i(\vec{k} \cdot \vec{p_j})}] 
\end{align}
\tag{7}
$$

At this point it can already be clearly seen how there is a restoring $A$-$A$ interaction even within the $A$-$B$ force term. This term can be defined as
$$
\begin{align}
\sum_{j=1}^3 P_j &= C_1 \sum_{j=1}^3(\hat{h_j}\hat{h_j}^T) + C_3 \sum_{j=1}^3(\hat{b_j}\hat{b_j}^T) \\
\sum_{j=1}^3 P_j &= 
\begin{pmatrix} 1 + \frac{1}{4} + \frac{1}{4} & 0 \\ 0 & 0 + \frac{3}{4} + \frac{3}{4} \end{pmatrix} = 
\frac{3}{2} \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} = \frac{3}{2}I(C_1 + C_3)
\end{align}
$$

#### Dynamic Matrices
The net forces for both the $A$-$A$ and $A$-$B$ interactions can now be fully described as
$$
A \text{-} A = \frac{3}{2}I(C_1 + C_3) + 
C_2 \left[\begin{pmatrix} \frac{3}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} (1 - e^{i(\vec{k} \cdot \vec{p}_{1,3})}) +
\begin{pmatrix} \frac{3}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} (1 - e^{i(\vec{k} \cdot \vec{p}_{2,4})}) +
\begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} (1 - e^{i(\vec{k} \cdot \vec{p}_{5,6})}) \right]
$$
$$
A \text{-} B = 
-C_1 \left[\begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} e^{i(\vec{k} \cdot \vec{h_1})} + 
\begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{h_2})} +
\begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{h_3})} \right]
-C_3 \left[\begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} e^{i(\vec{k} \cdot \vec{b_1})} + 
\begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{b_2})} +
\begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{b_3})} \right]
$$

Note how the leading terms in the $A$-$B$ interactions are all negative due to the swap around of the force equation  
The matrices describing all these forces take the form
$$
D^{AA} = \begin{bmatrix} D^{AA}_{xx} & D^{AA}_{xy} \\ D^{AA}_{yx} & D^{AA}_{yy} \end{bmatrix} \quad
\begin{align}
& D^{AA}_{xx} = \quad
\frac{3}{2}(C_1 + C_3) + C_2 [ \frac{3}{4}(2 - e^{i(\vec{k} \cdot \vec{p}_{1,3})} - e^{i(\vec{k} \cdot \vec{p}_{2,4})})] \\
& D^{AA}_{xy} \text{ and } D^{AA}_{yx} = \quad
C_2 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{p}_{2,4})} - e^{i(\vec{k} \cdot \vec{p}_{1,3})}) ] \\
& D^{AA}_{yy} = \quad
\frac{3}{2}(C_1 + C_3) + C_2 [ \frac{1}{4}(2 - e^{i(\vec{k} \cdot \vec{p}_{1,3})} - e^{i(\vec{k} \cdot \vec{p}_{2,4})}) + (1 - e^{i(\vec{k} \cdot \vec{p}_{5,6})}) ] \\
\end{align}
$$

$$
D^{AB} = \begin{bmatrix} D^{AB}_{xx} & D^{AB}_{xy} \\ D^{AB}_{yx} & D^{AB}_{yy} \end{bmatrix} \quad
\begin{align}
& D^{AB}_{xx} = \quad
-C_1 [ e^{i(\vec{k} \cdot \vec{h_1})} + \frac{1}{4}(e^{i(\vec{k} \cdot \vec{h_2})} + e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ e^{i(\vec{k} \cdot \vec{b_1})} + \frac{1}{4}(e^{i(\vec{k} \cdot \vec{b_2})} + e^{i(\vec{k} \cdot \vec{b_3})}) ] \\
& D^{AB}_{xy} \text{ and } D^{AB}_{yx} = \quad
-C_1 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{h_2})} - e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{b_3})} - e^{i(\vec{k} \cdot \vec{b_2})}) ] \\
& D^{AB}_{yy} = \quad
-C_1 [ \frac{3}{4}(e^{i(\vec{k} \cdot \vec{h_2})} + e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ \frac{3}{4}(e^{i(\vec{k} \cdot \vec{b_2})} + e^{i(\vec{k} \cdot \vec{b_3})}) ]
\end{align}
$$

**Please note that each p term represents $2$ vectors in the $\pm$ domain**  
So while the terms are displayed as $2 - e^{i(\vec{k} \cdot \vec{p}_{1,3})} - e^{i(\vec{k} \cdot \vec{p}_{2,4})}$ when expanded to $2cos(k p_j)$ each terms an extra $+2$ must be added since each term is linked as $(1-e^{i \vec{k} \cdot \vec{p}_j})$ making the full term $4 - cos(\vec{k} \cdot \vec{p}_1) - cos(\vec{k} \cdot \vec{p}_2)$.

$$
D^{AA} = \begin{bmatrix} D^{AA}_{xx} & D^{AA}_{xy} \\ D^{AA}_{yx} & D^{AA}_{yy} \end{bmatrix} \quad
\begin{align}
& D^{AA}_{xx} = \quad
\frac{3}{2}(C_1 + C_3) + C_2 [(3 - \frac{3}{2}(cos(\vec{k} \cdot \vec{p}_{1}) + cos(\vec{k} \cdot \vec{p}_{2}))) ] \\
& D^{AA}_{xy} \text{ and } D^{AA}_{yx} = \quad
C_2 [ \frac{\sqrt{3}}{2}( cos(\vec{k} \cdot \vec{p}_{2}) - cos(\vec{k} \cdot \vec{p}_{1})) ] \\
& D^{AA}_{yy} = \quad
\frac{3}{2}(C_1 + C_3) + C_2 [(3 - \frac{1}{2}(cos(\vec{k} \cdot \vec{p}_{1}) + cos(\vec{k} \cdot \vec{p}_{2})) - 2cos(\vec{k} \cdot \vec{p}_{5}) ] \\
\end{align}
$$

$$
D^{AB} = \begin{bmatrix} D^{AB}_{xx} & D^{AB}_{xy} \\ D^{AB}_{yx} & D^{AB}_{yy} \end{bmatrix} \quad
\begin{align}
& D^{AB}_{xx} = \quad
-C_1 [ e^{i(\vec{k} \cdot \vec{h_1})} + \frac{1}{4}(e^{i(\vec{k} \cdot \vec{h_2})} + e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ e^{i(\vec{k} \cdot \vec{b_1})} + \frac{1}{4}(e^{i(\vec{k} \cdot \vec{b_2})} + e^{i(\vec{k} \cdot \vec{b_3})}) ] \\
& D^{AB}_{xy} \text{ and } D^{AB}_{yx} = \quad
-C_1 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{h_2})} - e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{b_3})} - e^{i(\vec{k} \cdot \vec{b_2})}) ] \\
& D^{AB}_{yy} = \quad
-C_1 [ \frac{3}{4}(e^{i(\vec{k} \cdot \vec{h_2})} + e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ \frac{3}{4}(e^{i(\vec{k} \cdot \vec{b_2})} + e^{i(\vec{k} \cdot \vec{b_3})}) ]
\end{align}
$$

#### Full Dynamic Matrix

The Dynamic matrix is now defined as
$$
D(k) =
\begin{bmatrix}
D^{AA} & D^{AB} \\
D^{BA} & D^{BB}
\end{bmatrix}
=
\begin{bmatrix}
D^{AA}_{xx} & D^{AA}_{xy} & D^{AB}_{xx} & D^{AB}_{xy} \\
D^{AA}_{yx} & D^{AA}_{yy} & D^{AB}_{yx} & D^{AB}_{yy} \\
D^{BA}_{xx} & D^{BA}_{xy} & D^{BB}_{xx} & D^{BB}_{xy} \\
D^{BA}_{yx} & D^{BA}_{yy} & D^{BB}_{yx} & D^{BB}_{yy} 
\end{bmatrix}
$$

However the only values calculated were $D^{AB}$ and $D^{AA}$. This is fine due to the Hermitian symmetry of the Dynamic matrix. The term $D^{BB} = D^{AA}$ and the term $D^{BA} = D^{AB\dagger}$. Therefor the matrix can be decribed as 
$$
D(k) =
\begin{bmatrix}
D^{AA} & D^{AB} \\
D^{AB\dagger} & D^{AA}
\end{bmatrix}
$$

and used to solve
$$
\omega^2
\begin{bmatrix}
M_A & 0 & 0 & 0 \\
0 & M_A & 0 & 0 \\
0 & 0 & M_B & 0 \\
0 & 0 & 0 & M_B 
\end{bmatrix}
U_A
=
\begin{bmatrix}
D^{AA} & D^{AB} \\
D^{AB\dagger} & D^{AA}
\end{bmatrix}
U_A
\tag{8}
$$

Which is now the form of a standard Eigenvalue problem and can be evaluated with linear algebra solvers

#### High Symmetry Points
The high symmetry points represent the actions of a lattice at specific points. In the case of a hexagonal lattice there are three key symmetry points defined as $\Gamma, K, M$

These points define specific locations along the lattice within the Brillouin Zone, and as such are described by the reciprocal lattice vectors
$$
\begin{align}
b_1 = \frac{2\pi}{a} \begin{pmatrix} \frac{1}{3} \\ \frac{1}{\sqrt{3}} \end{pmatrix} \quad
b_2 = \frac{2\pi}{a} \begin{pmatrix} \frac{1}{3} \\ -\frac{1}{\sqrt{3}} \end{pmatrix}
\end{align}
$$

The $\Gamma$ point represents the origin of the BZ and as such sits as 
$$ \Gamma = \begin{pmatrix} 0 \\ 0 \end{pmatrix} $$
The $K$ point represents any particular corner of the BZ and as such there are 6 different variations. This $K$ is defined as 
$$ 
K = \frac{1}{3}(b_1,b_2) = \frac{2\pi}{3a} \begin{pmatrix} \frac{1}{3} \\ \frac{1}{3\sqrt{3}} \end{pmatrix}
$$
The $M$ point represents any particular midpoint of the BZ and there are also 6 different variations. This $M$ is defined as 
$$ 
M = \frac{1}{2}(b_1 + b_2) = \frac{2\pi}{a} \begin{pmatrix} \frac{1}{3} \\ 0 \end{pmatrix}
$$

### Diatomic (Honeycomb) Lattice

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt

# Physical parameters
mA = 1.0 # mass of A
mB = 1.0 # mass of B
C1 = 5.0
C2 = 2.4
C3 = 1.0
a = 1.0 # unit cell spacing

Nk = 1000

# Mass matrix
mass = np.zeros((4, 4))
np.fill_diagonal(mass, [mA, mA, mB, mB])

# Displacement Vectors
a1 = a*np.array([3/2 , np.sqrt(3)/2])
a2 = a*np.array([3/2 , -np.sqrt(3)/2])

h1 = a*np.array([1 , 0])
h2 = h1 - a1
h3 = h1 - a2

p1 = a1
p2 = a2
p5 = (a1 - a2)

b1 = -2*h1
b2 = h2 + a2
b3 = h3 + a1


# High-symmetry points
G = np.array([0.0, 0.0])
M = np.array([2*np.pi/(3*a) , 0.0])  # midpoint of BZ edge 
K = np.array([2*np.pi/(3*a), 2*np.pi/(3*np.sqrt(3)*a)])  # corner of BZ

k1 = np.linspace(G, K, Nk, endpoint=True)
k2 = np.linspace(K, M, Nk, endpoint=True)[1:]  # drop duplicate K
k3 = np.linspace(M, G, Nk, endpoint=True)[1:]  # drop duplicate M
k_path = np.vstack([k1, k2, k3])

k_dist = np.zeros(len(k_path))
for i in range(1, len(k_path)):
    dk = k_path[i] - k_path[i-1]
    k_dist[i] = k_dist[i-1] + np.linalg.norm(dk)

# Initialize eigenvalues
omega = np.zeros((len(k_path), 4))

# Looping through k-points to build force matrices
for i, (kx, ky) in enumerate(k_path):
    
    # Shorthand notation
    k = np.array([kx, ky])
    
    e = np.exp
    i_c = 1j
    s3 = np.sqrt(3)

    # Phase factors
    # =========================
    
    # h-vectors
    eh1 = e(i_c * np.dot(k, h1))
    eh2 = e(i_c * np.dot(k, h2))
    eh3 = e(i_c * np.dot(k, h3))
 
    # p-vectors (cosines of dot products)
    c13 = np.cos(np.dot(k, p1))   # = cos(k·p1)
    c24 = np.cos(np.dot(k, p2))   # = cos(k·p2)
    c56 = np.cos(np.dot(k, p5))   # = cos(k·p5)

    # b-vectors
    eb1 = e(i_c * np.dot(k, b1))
    eb2 = e(i_c * np.dot(k, b2))
    eb3 = e(i_c * np.dot(k, b3))
    
    # D_AA components
    D_AA_xx = (3/2*(C1 + C3) + C2*( 3 - (3/2)*(c13 + c24) ))
    D_AA_xy = (C2 * (s3/2) * (c24 - c13))
    D_AA_yx = D_AA_xy
    D_AA_yy = (3/2*(C1 + C3) + C2*( 3 - (1/2)*(c13 + c24) - 2*c56 ))
    
    D_AA = np.array([[D_AA_xx, D_AA_xy],
                     [D_AA_yx, D_AA_yy]])
    
    # D_AB components
    D_AB_xx = (
        -C1*(eh1 + 1/4*(eh2 + eh3)) -C3*(eb1 + 1/4*(eb2 + eb3)) )

    D_AB_xy = (
        -C1*(s3/4*(eh2 - eh3)) -C3*(s3/4*(eb3 - eb2)) )

    D_AB_yx = D_AB_xy

    D_AB_yy = (
        -C1*(3/4*(eh2 + eh3)) -C3*(3/4*(eb2 + eb3)) )

    D_AB = np.array([
        [D_AB_xx, D_AB_xy],
        [D_AB_yx, D_AB_yy] ])

    # Hermitian conjugate
    D_BA = D_AB.conj().T

    # Same-sublattice B
    D_BB = D_AA.copy()

    # Full matrix
    D = np.block([
        [D_AA, D_AB],
        [D_BA, D_BB]
    ])

    w2 = la.eigvalsh(D, mass)
    w2[w2 < 0] = 0.0
    omega[i] = np.sqrt(w2)


# Plotting
plt.figure(figsize=(8,5))

for n in range(4):
    if n < 2:
        plt.plot(k_dist, omega[:, n], 'b', label='Acoustic' if n == 0 else "")
    else:
        plt.plot(k_dist, omega[:, n], 'r', label='Optical' if n == 2 else "")

idxG = 0
idxK = Nk - 1
idxM = 2*Nk - 2

xticks = [k_dist[idxG], k_dist[idxK], k_dist[idxM], k_dist[-1]]
xtick_labels = [r'$\Gamma$', r'$K$', r'$M$', r'$\Gamma$']

plt.xticks(xticks, xtick_labels)
for t in xticks:
    plt.axvline(t, color='k', lw=0.5, alpha=0.3, linestyle='--')

plt.xlabel('Wave vector')
plt.ylabel(r'$\omega$')
plt.legend()
plt.tight_layout()
plt.show()
```

### Chirality

Chirality in chemistry refers to the mirrored symmetry of molecules. A molecule that can be perfectly mirrored is considered non-chiral, however in physics the chirality refers to the motion of particles. A chiral system is one in which the system reacts differently when mirrored creating a left-handed and right-handed form.  
Due to this change in action the act of rotating the system breaks time-reversal symmetry (TRS) as the reactions of the system are not persistent before and after changing handedness. To observe this change a small perturbation term can be introduced into calculations to represent changes caused by things such as Lorentz forces, Raman scattering, or Coriolis forces.  
The bond matrices are now represented as 
$$
\begin{align}
& \Phi(r_{n}) = \Phi_0 (r_{n}) + \Phi_{chiral} (r_{n}) \\
& \Phi(r_{n}) = \Phi_0 (r_{n}) + i \gamma \frac{r_{n,x}}{|r_{n}|} S_z
\end{align}
\tag{9}
$$

The term $\gamma$ is a dimensionless gyroscopic coupling parameter that exists as any value $ \gamma > 0$, but typically on a scale of the bond coefficients in that system. The directional vectors in $\hat{x}$ and $S_z$ represent the time reversal breaking impact on neighboring interactions with a transformation only along the x-axis where
$$
S_z =
\begin{bmatrix}
0 & -1 \\ 
1 & 0
\end{bmatrix} 
\quad
\frac{r_{ij,x}}{|r_{ij}|} = cos(\theta_{ij})
$$

#### The new Dynamic Matrix

The standard Dynamic matrix is formed by Bloch's theorem as the form
$$ 
D(k) = \sum_n \Phi_0 e^{i \vec{k} \cdot r_n}
$$

So by introducing $Eq9$ as the new bond matrix this becomes
$$
D(k) = \sum_n \Phi_0 e^{i \vec{k} \cdot r_n} + ig \hat{r}_{n,x} S_2 e^{i \vec{k} \cdot r_n} = D_0(k) + igG(k)
\tag{10}
$$

By using the predefined neighbor vectors the x-component can be extracted and the $A-B$ interactions with $h_n$ and $b_n$ terms can be used while the $A-A$ interactions rely primarily on $p_n$ terms. This gives the perturbation terms

$$
\begin{align}
& AB: 
ig \left[ (e^{i \vec{k} \cdot \vec{h_1}} - \frac{1}{2}(e^{i \vec{k} \cdot \vec{h_2}} + e^{i \vec{k} \cdot \vec{h_3}})) -
(e^{i \vec{k} \cdot \vec{b_1}} - \frac{1}{2}(e^{i \vec{k} \cdot \vec{b_2}} + e^{i \vec{k} \cdot \vec{b_3}})) \right] 
\begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix} \\
& AA:
ig \left[ \frac{\sqrt{3}}{2}(e^{i \vec{k} \cdot \vec{p_1}} + e^{i \vec{k} \cdot \vec{p_2}}) - \frac{\sqrt{3}}{2}(e^{i \vec{k} \cdot \vec{p_3}} + e^{i \vec{k} \cdot \vec{p_4}}) \right] 
\begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}
\end{align}
$$

When including a TRS-breaking perturbation for honeycomb phonons, only the nearest-neighbor (A–B, h) bonds are actually included in the antisymmetric term. The second and third-nearest neighbors are excluded. For the second-nearest neighbor bonds (A-A B-B, p) this is because the chiral motion is dominated by the interatomic bonds between atoms in the basis, and as such intra-sublattice terms can be neglected. Including them breaks the degeneracy of the system when determining the eigenvalues and leads to unrealistic solutions. For the third-nearest neighbor terms (A-B, b) while these are still terms describing atomic bonds they are not within the same unit cell, but furthermore the unit vectors for $\vec{h}$ and $\vec{b}$ are exact opposites, so by symmetry they cancel any present perturbations when both included in the sum.

### (Pseudo-) Angular Momentum

With these new perturbations there are now breaks along the K points of the Brillouin Zone (BZ). These breaks arise due to the new lack of degeneracy from the perturbation term, one branch will oscillate counter-clockwise (right handed) and one branch will oscillate clockwise (left handed). For each wave vector $k$ the solutions to the Dynamic matrix have eigenvectors
$$
D(k) e_{nk} = \omega^2_{nk} e_{nk}
\tag{11}
$$
where $n$ is the band index for those solutions.  
The angular momentum is defined as
$$
L^z_{nk} = \hbar \sum_{s,\alpha,\beta} \epsilon^{z \alpha \beta} e^*_{ns \alpha}(k) e_{ns \beta}(k)
\tag{12}
$$
Where $\epsilon$ is the Levi-Civita tensor and the $e$ values represent the eigenvector for a unit cell $s$ with positions $(x,y,z) = (\alpha,\beta,z)$.  
For a planar lattice this can be simplified to 
$$
L^z_{nk} = \hbar \sum_s  e^*_{nsx}(k) e_{nsy}(k) - e^*_{nsy}(k) e_{nsx}(k)
\tag{13}
$$
Which further simplifies to the common 2D form
$$
L^z_{nk} = 2\hbar \sum_s  Im(e^*_{nsx} e_{nsy})
\tag{14}
$$
Where $L^z > 0$ is right handed (counter-clockwise) and $L^z < 0$ is left handed (clockwise) motion.  

Why exactly are there chiral states? And why at the K points?  
In the hexagonal lattice the phonon modes are free to oscillate in the $\hat{x}$ and $\hat{y}$ directions independently, but due to the periodicity of the lattice and how the atoms bond any motion the phonons have must be rotationally invariant. If the honeycomb lattice were rotated $120 \degree$ then the motion of phonons at these points can no longer be independent in $\hat{x}$ and $\hat{y}$ directions, but rather a linear combination of each. As such the vibrational modes seek out a motion that is invariant which just so happens to be circular motion with $K$ point values being right handed and $K' (-K)$ points being left handed. These vibrational modes are not exclusive to the K points, but in many other cases the chirality is not required for symmetry defined by the lattice it just happens to occur.

### Density of States

The Density of States (DoS) at a given frequency is given by
$$
D(\omega) = \frac{1}{N_k} \sum_{n=1}^{N_{bands}} \sum_k \delta (\omega - \omega_n(k))
\tag{15}
$$

This can be broken down into the projeced DoS by introducing a term $e_{n,\alpha}(k)$ to represent the eigenvector describing each specific degree of freedom for a mass at each wave vector, $k$. 
$$
D_{\alpha}(\omega) = \frac{1}{N_k} \sum_{n=1}^{N_{bands}} \sum_k |e_{n,\alpha}(k)|^2 \delta (\omega - \omega_n(k))
\tag{16}
$$

This form however creates very descrete boxes for DoS, for a clearly visible flow the Dirac-delta function is replaced by a smooth Gaussian curve
$$
\delta (\omega - \omega_n(k)) = \frac{1}{\sqrt{2 \pi} \sigma} exp{ -\frac{\omega - \omega_n(k)}{2 \sigma^2}}
$$
where each curve is now centered on $\omega_n(k)$ with a smoothness adjusted by $\sigma$ which gives the form 
$$
D_{\alpha}(\omega) = \frac{1}{N_k} \sum_{n=1}^{N_{bands}} \sum_k |e_{n,\alpha}(k)|^2 \frac{1}{\sqrt{2 \pi} \sigma} exp{ -\frac{\omega - \omega_n(k)}{2 \sigma^2}}
\tag{17}
$$

So by finding the explicite frequency $\omega_n$ as the eigenvalues from the Dynamic matrix and normalizing by $\sqrt{2\pi}\sigma N_k$ to preserve the totaly area then each DoS can be determined for the different masses.

###  Γ → K → M → Γ Path

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt

# Physical parameters
mA = 1.0 # mass of A
mB = 1.0 # mass of B
C1 = 5.0
C2 = 2.0
C3 = 1.0
a = 1.0 # unit cell spacing
g = 0.37 * C1 # gyroscopic parameter

Nk = 1000

# Mass matrix
mass = np.zeros((4, 4))
np.fill_diagonal(mass, [mA, mA, mB, mB])

# Angular momentum 
def mass_normalize(e, mass):
    norm = np.real(np.conj(e) @ (mass @ e))
    return e / np.sqrt(norm)

def compute_Lz(e):
    # basis: (x_A, y_A, x_B, y_B)
    eAx, eAy, eBx, eBy = e
    
    Lz_A = 2 * np.imag(np.conj(eAx) * eAy)
    Lz_B = 2 * np.imag(np.conj(eBx) * eBy)
    
    return Lz_A + Lz_B
    
# Displacement Vectors
a1 = a*np.array([3/2 , np.sqrt(3)/2])
a2 = a*np.array([3/2 , -np.sqrt(3)/2])

h1 = a*np.array([1 , 0])
h2 = h1 - a1
h3 = h1 - a2

p1 = a1
p2 = a2
p5 = (a1 - a2)

b1 = -2*h1
b2 = h2 + a2
b3 = h3 + a1


# High-symmetry points
G = np.array([0.0, 0.0])
M = np.array([2*np.pi/(3*a) , 0.0])  # midpoint of BZ edge 
K = np.array([2*np.pi/(3*a), 2*np.pi/(3*np.sqrt(3)*a)])  # corner of BZ

# Standard k path
k1 = np.linspace(G, K, Nk, endpoint=True)
k2 = np.linspace(K, M, Nk, endpoint=True)[1:]  # drop duplicate K
k3 = np.linspace(M, G, Nk, endpoint=True)[1:]  # drop duplicate M
k_path = np.vstack([k1, k2, k3])

k_dist = np.zeros(len(k_path))
for i in range(1, len(k_path)):
    dk = k_path[i] - k_path[i-1]
    k_dist[i] = k_dist[i-1] + np.linalg.norm(dk)

# Initialize eigenvalues and angular momentum
omega = np.zeros((len(k_path), 4))
Lz = np.zeros((len(k_path), 4))

# Looping through k-points to build force matrices
for i, (kx, ky) in enumerate(k_path):
    
    # Shorthand notation
    k = np.array([kx, ky])
    
    e = np.exp
    i_c = 1j
    s3 = np.sqrt(3)

    # Phase factors
    # h-vectors
    eh1 = e(i_c * np.dot(k, h1))
    eh2 = e(i_c * np.dot(k, h2))
    eh3 = e(i_c * np.dot(k, h3))
 
    # p-vectors (cosines of dot products)
    c13 = np.cos(np.dot(k, p1))   # = cos(k·p1)
    c24 = np.cos(np.dot(k, p2))   # = cos(k·p2)
    c56 = np.cos(np.dot(k, p5))   # = cos(k·p5)

    # b-vectors
    eb1 = e(i_c * np.dot(k, b1))
    eb2 = e(i_c * np.dot(k, b2))
    eb3 = e(i_c * np.dot(k, b3))
    
    # D_AA components
    D_AA_xx = (3/2*(C1 + C3) + C2*( 3 - (3/2)*(c13 + c24) ))
    D_AA_xy = (C2 * (s3/2) * (c24 - c13)) 
    D_AA_yx = D_AA_xy
    D_AA_yy = (3/2*(C1 + C3) + C2*( 3 - (1/2)*(c13 + c24) - 2*c56 ))
    
    D_AA = np.array([[D_AA_xx, D_AA_xy],
                     [D_AA_yx, D_AA_yy]])
    
    # D_AB components
    D_AB_xx = (
        -C1*(eh1 + 1/4*(eh2 + eh3)) -C3*(eb1 + 1/4*(eb2 + eb3)) )

    D_AB_xy = (
        -C1*(s3/4*(eh2 - eh3)) -C3*(s3/4*(eb3 - eb2)) )

    D_AB_yx = D_AB_xy

    D_AB_yy = (
        -C1*(3/4*(eh2 + eh3)) -C3*(3/4*(eb2 + eb3)) )

    D_AB = np.array([
        [D_AB_xx, D_AB_xy],
        [D_AB_yx, D_AB_yy] ])

    # Creating antisymmetric matrix ###########################
    # Chiral (TRS-breaking) terms
    Sz = np.array([[0, -1],
                   [1,  0]])
    
    # ---- AB chiral term ----
    G_AB = ( (eh1 - 0.5*(eh2 + eh3))   # h bonds
       # - (eb1 - 0.5*(eb2 + eb3)) # b bonds 
           )
    
    D_AB_chiral = 1j * g * G_AB * Sz
    '''
    # ---- AA chiral term ----
    # Need exponentials for p-vectors (not cosines)
    ep1 = np.exp(1j * np.dot(k, p1))
    ep2 = np.exp(1j * np.dot(k, p2))
    ep3 = np.exp(-1j * np.dot(k, p1))  # -p1
    ep4 = np.exp(-1j * np.dot(k, p2))  # -p2
    # p5/p6 give zero (x-component = 0), so omitted
    
    G_AA = (np.sqrt(3)/2) * (
        (ep1 + ep2) - (ep3 + ep4))
    
    D_AA_chiral = 1j * g * G_AA * S2
    '''
    # Add antisymmetric matrix to Dynamic matrix
    D_AB = D_AB + D_AB_chiral
    #D_AA = D_AA + D_AA_chiral

    # Hermitian conjugate
    D_BA = D_AB.conj().T

    # Same-sublattice B
    D_BB = D_AA.copy()

    # Full matrix
    D = np.block([
        [D_AA, D_AB],
        [D_BA, D_BB] ])

    w2, vecs = la.eigh(D, mass)
    w2[w2 < 0] = 0.0
    omega[i] = np.sqrt(w2)

    # computing Lz
    for n in range(4):
        e = mass_normalize(vecs[:, n], mass)
        
        # optional phase fix (keeps consistency)
        phase = np.angle(e[0])
        e = e * np.exp(-1j * phase)
        
        Lz_val = compute_Lz(e)
        Lz[i, n] = Lz_val
        
       
plt.figure(figsize=(10,7))

for n in range(4):
    if n < 2:
        plt.plot(k_dist, omega[:, n], 'b', label='Acoustic' if n == 0 else "")
    else:
        plt.plot(k_dist, omega[:, n], 'r', label='Optical' if n == 2 else "")

idxG = 0
idxK = Nk - 1
idxM = 2*Nk - 2

xticks = [k_dist[idxG], k_dist[idxK], k_dist[idxM], k_dist[-1]]
xtick_labels = [r'$\Gamma$', r'$K$', r'$M$', r'$\Gamma$']

plt.xticks(xticks, xtick_labels)
for t in xticks:
    plt.axvline(t, color='k', lw=0.5, alpha=0.3, linestyle='--')
    
# circular arrows for chiral motion    
offset = 5
idx = idxK + offset
print("\nLz values used for plotting:")
for n in range(4):
    val = Lz[idx, n]
    
    if val > 0:
        hand = "right-handed"
    elif val < 0:
        hand = "left-handed"
    else:
        hand = "non-chiral"
    print(f"band {n}: Lz = {val:.6f} ({hand})")
    
for n in range(4):
    val = 0.5 * (Lz[idxK + offset, n] + Lz[idxK - offset, n])
    
    if abs(val) < 1e-4:
        continue
    
    x = k_dist[idxK]
    y = omega[idxK, n]
    
    marker = '↺' if val > 0 else '↻'
    
    plt.text(x, y, marker, fontsize=18,
             ha='center', va='center', color='k')


plt.xlabel('Wave vector')
plt.ylabel(r'Frequency ($\omega$)')
plt.legend()
plt.tight_layout()
plt.show()
```

###  K' → Γ → K Path

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt

# Physical parameters
mA = 1.0 # mass of A
mB = 1.0 # mass of B
C1 = 5.0
C2 = 2.0
C3 = 1.0
a = 1.0 # unit cell spacing
g = 0.37 * C1 # gyroscopic parameter

Nk = 1000

# Mass matrix
mass = np.zeros((4, 4))
np.fill_diagonal(mass, [mA, mA, mB, mB])

# Displacement Vectors
a1 = a*np.array([3/2 , np.sqrt(3)/2])
a2 = a*np.array([3/2 , -np.sqrt(3)/2])

h1 = a*np.array([1 , 0])
h2 = h1 - a1
h3 = h1 - a2

p1 = a1
p2 = a2
p5 = (a1 - a2)

b1 = -2*h1
b2 = h2 + a2
b3 = h3 + a1


# High-symmetry points
G = np.array([0.0, 0.0])
K = np.array([2*np.pi/(3*a), 2*np.pi/(3*np.sqrt(3)*a)])  # corner of BZ
K_prime = np.array([0.0, 2*np.pi/(3*a)]) # top corner of BZ

eps = 0.25 # extension factor to observe slightly past K points

# Direction vector along K
Kp_dir = K_prime / np.linalg.norm(K_prime)

# Extended endpoints
Kp_ext = K_prime + eps * Kp_dir
Kpneg_ext = -K_prime - eps * Kp_dir

# New path: (-K - ε) → -K → Γ → K → (K + ε)
k0 = np.linspace(Kpneg_ext, -K_prime, Nk//5, endpoint=False)
k1 = np.linspace(-K_prime, G, Nk, endpoint=False)
k2 = np.linspace(G, K_prime, Nk, endpoint=False)
k3 = np.linspace(K_prime, Kp_ext, Nk//5, endpoint=True)

k_path = np.vstack([k0, k1, k2, k3])

k_dist = np.zeros(len(k_path))
for i in range(1, len(k_path)):
    dk = k_path[i] - k_path[i-1]
    k_dist[i] = k_dist[i-1] + np.linalg.norm(dk)

# Initialize eigenvalues
omega = np.zeros((len(k_path), 4))

# Looping through k-points to build force matrices
for i, (kx, ky) in enumerate(k_path):
    
    # Shorthand notation
    k = np.array([kx, ky])
    
    e = np.exp
    i_c = 1j
    s3 = np.sqrt(3)

    # Phase factors
    # h-vectors
    eh1 = e(i_c * np.dot(k, h1))
    eh2 = e(i_c * np.dot(k, h2))
    eh3 = e(i_c * np.dot(k, h3))
 
    # p-vectors (cosines of dot products)
    c13 = np.cos(np.dot(k, p1))   # = cos(k·p1)
    c24 = np.cos(np.dot(k, p2))   # = cos(k·p2)
    c56 = np.cos(np.dot(k, p5))   # = cos(k·p5)

    # b-vectors
    eb1 = e(i_c * np.dot(k, b1))
    eb2 = e(i_c * np.dot(k, b2))
    eb3 = e(i_c * np.dot(k, b3))
    
    # D_AA components
    D_AA_xx = (3/2*(C1 + C3) + C2*( 3 - (3/2)*(c13 + c24) ))
    D_AA_xy = (C2 * (s3/2) * (c24 - c13)) 
    D_AA_yx = D_AA_xy
    D_AA_yy = (3/2*(C1 + C3) + C2*( 3 - (1/2)*(c13 + c24) - 2*c56 ))
    
    D_AA = np.array([[D_AA_xx, D_AA_xy],
                     [D_AA_yx, D_AA_yy]])
    
    # D_AB components
    D_AB_xx = (
        -C1*(eh1 + 1/4*(eh2 + eh3)) -C3*(eb1 + 1/4*(eb2 + eb3)) )

    D_AB_xy = (
        -C1*(s3/4*(eh2 - eh3)) -C3*(s3/4*(eb3 - eb2)) )

    D_AB_yx = D_AB_xy

    D_AB_yy = (
        -C1*(3/4*(eh2 + eh3)) -C3*(3/4*(eb2 + eb3)) )

    D_AB = np.array([
        [D_AB_xx, D_AB_xy],
        [D_AB_yx, D_AB_yy] ])

    # Creating antisymmetric matrix ###########################
    # Chiral (TRS-breaking) terms
    Sz = np.array([[0, -1],
                   [1,  0]])
    
    # ---- AB chiral term ----
    G_AB = ( (eh1 - 0.5*(eh2 + eh3))   # h bonds
            )
    
    D_AB_chiral = 1j * g * G_AB * Sz

    # Add antisymmetric matrix to Dynamic matrix
    D_AB = D_AB + D_AB_chiral

    # Hermitian conjugate
    D_BA = D_AB.conj().T

    # Same-sublattice B
    D_BB = D_AA.copy()

    # Full matrix
    D = np.block([
        [D_AA, D_AB],
        [D_BA, D_BB] ])

    w2 = la.eigvalsh(D, mass)
    w2[w2 < 0] = 0.0
    omega[i] = np.sqrt(w2)
    

# Plotting
plt.figure(figsize=(8,5))

for n in range(4):
    if n < 2:
        plt.plot(k_dist, omega[:, n], 'b', label='Acoustic' if n == 0 else "")
    else:
        plt.plot(k_dist, omega[:, n], 'r', label='Optical' if n == 2 else "")

idx_Kneg = len(k0)
idx_G = len(k0) + len(k1)
idx_Kp = len(k0) + len(k1) + len(k2)

xticks = [k_dist[idx_Kneg], k_dist[idx_G], k_dist[idx_Kp]]
xtick_labels = [r"$K'$", r'$\Gamma$', r"$K'$"]

plt.xticks(xticks, xtick_labels)
for t in xticks:
    plt.axvline(t, color='k', lw=0.5, alpha=0.3, linestyle='--')
    
plt.xlim(k_dist[0], k_dist[-1])
plt.xlabel('Wave vector')
plt.ylabel(r'Frequency ($\omega$)')
plt.legend()
plt.tight_layout()
plt.show()
```

### DoS

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt

# Physical parameters
mA = 1.0
mB = 1.0
C1 = 5.0
C2 = 2.0
C3 = 1.0
a = 1.0
g = 0.37 * C1

Nk = 1000

# Mass matrix
mass = np.zeros((4, 4))
np.fill_diagonal(mass, [mA, mA, mB, mB])

# Displacement Vectors
a1 = a*np.array([3/2 , np.sqrt(3)/2])
a2 = a*np.array([3/2 , -np.sqrt(3)/2])

h1 = a*np.array([1 , 0])
h2 = h1 - a1
h3 = h1 - a2

p1 = a1
p2 = a2
p5 = (a1 - a2)

b1 = -2*h1
b2 = h2 + a2
b3 = h3 + a1

# High-symmetry points
G = np.array([0.0, 0.0])
M = np.array([2*np.pi/(3*a) , 0.0])
K = np.array([2*np.pi/(3*a), 2*np.pi/(3*np.sqrt(3)*a)])

# k path
k1 = np.linspace(G, K, Nk, endpoint=True)
k2 = np.linspace(K, M, Nk, endpoint=True)[1:]
k3 = np.linspace(M, G, Nk, endpoint=True)[1:]
k_path = np.vstack([k1, k2, k3])

# k distance
k_dist = np.zeros(len(k_path))
for i in range(1, len(k_path)):
    dk = k_path[i] - k_path[i-1]
    k_dist[i] = k_dist[i-1] + np.linalg.norm(dk)

# Storage
omega = np.zeros((len(k_path), 4))

# -------- DoS SETUP --------
omega_max_est = 5.0
omega_grid = np.linspace(0, omega_max_est, 500)
sigma = 0.03

dos_total = np.zeros_like(omega_grid)
dos_proj = np.zeros((4, len(omega_grid)))
# --------------------------

# Loop over k
for i, (kx, ky) in enumerate(k_path):

    k = np.array([kx, ky])
    e = np.exp
    i_c = 1j
    s3 = np.sqrt(3)

    eh1 = e(i_c * np.dot(k, h1))
    eh2 = e(i_c * np.dot(k, h2))
    eh3 = e(i_c * np.dot(k, h3))

    c13 = np.cos(np.dot(k, p1))
    c24 = np.cos(np.dot(k, p2))
    c56 = np.cos(np.dot(k, p5))

    eb1 = e(i_c * np.dot(k, b1))
    eb2 = e(i_c * np.dot(k, b2))
    eb3 = e(i_c * np.dot(k, b3))

    # D_AA
    D_AA = np.array([
        [3/2*(C1+C3) + C2*(3 - 3/2*(c13+c24)),
         C2*(s3/2)*(c24-c13)],
        [C2*(s3/2)*(c24-c13),
         3/2*(C1+C3) + C2*(3 - 1/2*(c13+c24) - 2*c56)]
    ])

    # D_AB
    D_AB = np.array([
        [-C1*(eh1 + 1/4*(eh2+eh3)) - C3*(eb1 + 1/4*(eb2+eb3)),
         -C1*(s3/4*(eh2-eh3)) - C3*(s3/4*(eb3-eb2))],
        [-C1*(s3/4*(eh2-eh3)) - C3*(s3/4*(eb3-eb2)),
         -C1*(3/4*(eh2+eh3)) - C3*(3/4*(eb2+eb3))]
    ])

    # Chiral term
    Sz = np.array([[0, -1],
                   [1,  0]])

    G_AB = (eh1 - 0.5*(eh2 + eh3))
    D_AB += 1j * g * G_AB * Sz

    D_BA = D_AB.conj().T
    D_BB = D_AA.copy()

    D = np.block([[D_AA, D_AB],
                  [D_BA, D_BB]])

    # Eigenproblem
    w2, vecs = la.eigh(D, mass)
    w2[w2 < 0] = 0.0
    w = np.sqrt(w2)
    omega[i] = w

    # -------- DoS ACCUMULATION --------
    for n in range(4):
        wn = w[n]
        gauss = np.exp(-(omega_grid - wn)**2 / (2*sigma**2))

        dos_total += gauss

        for alpha in range(4):
            weight = np.abs(vecs[alpha, n])**2
            dos_proj[alpha] += weight * gauss
    # ---------------------------------

# Normalize DoS
norm = np.sqrt(2*np.pi)*sigma*len(k_path)
dos_total /= norm
dos_proj /= norm

# -------- BAND STRUCTURE --------
plt.figure(figsize=(8,5))

for n in range(4):
    if n < 2:
        plt.plot(k_dist, omega[:, n], 'b', label='Acoustic' if n == 0 else "")
    else:
        plt.plot(k_dist, omega[:, n], 'r', label='Optical' if n == 2 else "")

idxG = 0
idxK = Nk - 1
idxM = 2*Nk - 2

xticks = [k_dist[idxG], k_dist[idxK], k_dist[idxM], k_dist[-1]]
xtick_labels = [r'$\Gamma$', r'$K$', r'$M$', r'$\Gamma$']

plt.xticks(xticks, xtick_labels)
for t in xticks:
    plt.axvline(t, color='k', lw=0.5, alpha=0.3, linestyle='--')

plt.xlabel('Wave vector')
plt.ylabel(r'Frequency ($\omega$)')
plt.legend()
plt.tight_layout()
plt.show()

# -------- DoS PLOT --------
plt.figure(figsize=(5,6))

labels = [r'$A_x$', r'$A_y$', r'$B_x$', r'$B_y$']

for i in range(4):
    plt.plot(dos_proj[i], omega_grid,  label=labels[i])
    plt.fill_betweenx(omega_grid, 0, dos_proj[i], alpha=0.3)

plt.plot(dos_total, omega_grid, 'k', linewidth=2, label='Total')

plt.xlabel('DoS')
plt.ylabel(r'$\omega$')
plt.legend()
# Add bond info as text on the figure
plt.text(0.95, 0.05, f'C1 = {C1}\nC2 = {C2}\nC3 = {C3}', 
         transform=plt.gca().transAxes, 
         fontsize=10, 
         verticalalignment='bottom', 
         horizontalalignment='right',
         bbox=dict(facecolor='white', alpha=0.6, edgecolor='none'))

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation


# -----------------------------
# Honeycomb lattice
# -----------------------------
def generate_honeycomb(nx=10, ny=8, a=1.0):
    a1 = np.array([np.sqrt(3)*a, 0])
    a2 = np.array([np.sqrt(3)/2*a, 3/2*a])

    pos, sub = [], []

    for j in range(ny):
        for i in range(nx):
            R = i*a1 + j*a2
            pos.append(R + [0,0]); sub.append(0)
            pos.append(R + [0,a]); sub.append(1)

    return np.array(pos), np.array(sub)


# -----------------------------
# Bonds
# -----------------------------
def get_bonds(pos, cutoff=1.1):
    bonds = []
    for i in range(len(pos)):
        for j in range(i+1, len(pos)):
            if np.linalg.norm(pos[i] - pos[j]) < cutoff:
                bonds.append((i, j))
    return bonds


# -----------------------------
# GLOBAL PHONON MODE (NO PACKET)
# -----------------------------
def displacement(pos, sub, t, mode):
    k_dir = np.array([1.0, 0.0])
    k_dir /= np.linalg.norm(k_dir)

    omega = 2.5
    k = 0.8

    phase = pos @ k_dir * k - omega * t

    disp = np.zeros_like(pos)
    # -----------------------------
    # CHIRAL (TRUE circular polarization)
    # -----------------------------
    if mode == "chiral":
        disp = np.zeros_like(pos)
    
        # ONLY red atoms move
        mask = (sub == 0)
    
        k_dir = np.array([1.0, 0.0])
        k_dir /= np.linalg.norm(k_dir)
    
        k = 0.8
        omega = 2.5
    
        phase = pos @ k_dir * k - omega * t
    
        # chiral rotation ONLY on red atoms
        disp[mask, 0] = np.cos(phase[mask])
        disp[mask, 1] = np.sin(phase[mask])

    return disp


# -----------------------------
# Animation
# -----------------------------
def make_animation(mode, filename):
    pos, sub = generate_honeycomb()
    bonds = get_bonds(pos)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

    for ax in (ax1, ax2):
        ax.set_aspect('equal')
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_xlim(pos[:,0].min()-1, pos[:,0].max()+1)
        ax.set_ylim(pos[:,1].min()-1, pos[:,1].max()+1)

    # -----------------------------
    # bonds FIRST (behind everything)
    # -----------------------------
    lines1, lines2 = [], []

    for i, j in bonds:
        l1, = ax1.plot(
            [pos[i,0], pos[j,0]],
            [pos[i,1], pos[j,1]],
            'k-', lw=0.8, alpha=0.6, zorder=1
        )

        l2, = ax2.plot(
            [pos[i,0], pos[j,0]],
            [pos[i,1], pos[j,1]],
            'k-', lw=0.8, alpha=0.6, zorder=1
        )

        lines1.append(l1)
        lines2.append(l2)

    # -----------------------------
    # atoms (both sublattices)
    # -----------------------------
    scat1_A = ax1.scatter([], [], c='red', s=35, zorder=3)
    scat1_B = ax1.scatter([], [], c='blue', s=35, zorder=3)

    scat2_A = ax2.scatter([], [], c='red', s=35, zorder=3)
    scat2_B = ax2.scatter([], [], c='blue', s=35, zorder=3)

    ax1.set_title("Chiral (L)")
    ax2.set_title("Chiral (R)")

    def update(frame):
        t = frame * 0.15

        k_dir = np.array([1.0, 0.0])
        k_dir /= np.linalg.norm(k_dir)

        k = 0.8
        omega = 2.5

        phase = pos @ k_dir * k - omega * t

        red = (sub == 0)
        blue = (sub == 1)

        # -----------------------------
        # LEFT (original chirality)
        # -----------------------------
        disp_L = np.zeros_like(pos)
        disp_L[red,0] = np.cos(phase[red])
        disp_L[red,1] = np.sin(phase[red])

        new_L = pos + 0.08 * disp_L

        scat1_A.set_offsets(new_L[red])
        scat1_B.set_offsets(new_L[blue])

        # -----------------------------
        # RIGHT (opposite chirality)
        # -----------------------------
        disp_R = np.zeros_like(pos)
        disp_R[red,0] = np.cos(-phase[red])
        disp_R[red,1] = np.sin(-phase[red])

        new_R = pos + 0.08 * disp_R

        scat2_A.set_offsets(new_R[red])
        scat2_B.set_offsets(new_R[blue])

        # -----------------------------
        # update bonds
        # -----------------------------
        for idx, (i, j) in enumerate(bonds):
            lines1[idx].set_data([new_L[i,0], new_L[j,0]],
                                 [new_L[i,1], new_L[j,1]])

            lines2[idx].set_data([new_R[i,0], new_R[j,0]],
                                 [new_R[i,1], new_R[j,1]])

        return scat1_A, scat1_B, scat2_A, scat2_B, *lines1, *lines2

    ani = FuncAnimation(fig, update, frames=200, interval=50)
    ani.save(filename, writer='pillow', fps=20)
    plt.close()

# -----------------------------
# Run
# -----------------------------
make_animation("chiral", "chiral.gif")
```
