## Standard Derivation of 2D Lattice

### Monoatomic Basis

For a square lattice with atoms of mass $M$ there will be phonons in both the horizontal and vertical axis for each atom in the crystal. As such the plane waves describing the system must be contained within a matrix of the form

$$
\overrightarrow{U_n} =
\begin{bmatrix} 
U_{nm,x} (t) \\
U_{nm,y} (t)
\end{bmatrix}
=
\large
\begin{bmatrix}
A_x \\
A_y 
\end{bmatrix}
e^{i(\vec{k} \cdot \vec{R_{nm}} -\omega t)}
\tag{1}
$$

in this case the vector, 
$$
\vec{k} =
\begin{bmatrix}
k_x \\
ky
\end{bmatrix}
$$
contains the allowed wave vectors, and the vector,
$$
\overrightarrow{R_{nm}} =
\begin{bmatrix}
n \cdot a_x \\
m \cdot a_y 
\end{bmatrix}
$$
contains the unit cell spacing of atoms within the crystal.

The displacement between any two points on the lattice can be described by the vector
\begin{equation}
\vec{h} = \overrightarrow{U_{n',m'}} - \overrightarrow{U_{n,m}}
\end{equation}

which has a direction of
\begin{equation}
\hat{h} = \frac{\vec{h}}{||\vec{h}||}
\end{equation}

The displacements in a square lattice depend on the position of the nearest neighbors. The first nearest neighbors will exist on the horizontal and vertical planes such that 
$$
\begin{aligned}
h_1 &= a\hat{x} \quad
h_2 = a\hat{y} \\
h_3 &= -a\hat{x} \quad
h_4 = -a\hat{y}
\end{aligned}
$$
While the second nearest neighbors exist along the diagonals with displacements
$$
\begin{aligned}
p_1 &= a\hat{x} + a\hat{y} \quad
p_2 = -a\hat{x} + a\hat{y} \\
p_3 &= -a\hat{x} - a\hat{y} \quad
p_4 = a\hat{x} - a\hat{y}
\end{aligned}
$$

The directions corresponding to each of these vectors can be solved out such that  
$$
\begin{aligned}
\hat{h_1} &= \hat{x} \quad
\hat{h_2} = \hat{y} \\
\hat{h_3} &= -\hat{x} \quad
\hat{h_4} = -\hat{y}
\end{aligned}
$$
$$
\begin{aligned}
\hat{p_1} &= \frac{1}{\sqrt{2}}(\hat{x} + \hat{y}) \quad
\hat{p_2} = \frac{1}{\sqrt{2}}(-\hat{x} + \hat{y}) \\
\hat{p_3} &= \frac{1}{\sqrt{2}}(-\hat{x} - \hat{y})\quad
\hat{p_4} = \frac{1}{\sqrt{2}}(\hat{x} - \hat{y})
\end{aligned}
$$

Note: $h_j$ and $p_j$ represent the same variable and are just different for ease of seeing which belongs to 1st or 2nd nearest neighbors

Similar to the systems for 1D photons the forces can be solved with Newton's 2nd Law in the general form

\begin{equation}
M \frac{\delta^2 \overrightarrow{U_{nm}}}{\delta t^2} = C_1\sum_{j=1,2,3,4} [(\overrightarrow{U_{n',m'}} - \overrightarrow{U_{n,m}}) \cdot \hat{h_j}]\hat{h_j} + C_2\sum_{j=1,2,3,4} [(\overrightarrow{U_{n',m'}} - \overrightarrow{U_{n,m}}) \cdot \hat{p_j}]\hat{p_j}  + C_3....
\tag{2}
\end{equation}
Where the equation can be expanded to include as many nearest neighbors as required. In this square lattice only up to 2nd nearest is required.    

Centering on site $(n,m)$ this form can be altered to 
\begin{equation}
M \frac{\delta^2 \overrightarrow{U_{nm}}}{\delta t^2} = C_1\sum_{j=1,2,3,4} [(\overrightarrow{U_{n+h_j}} - \overrightarrow{U_{n}}) \cdot \hat{h_j}]\hat{h_j} + C_2\sum_{j=1,2,3,4} [(\overrightarrow{U_{n+p_j}} - \overrightarrow{U_{n}}) \cdot \hat{p_j}]\hat{p_j} 
\tag{3}
\end{equation}

And the plane wave solutions from $Eq1$ can be called into the problem using the solution that
$$
\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}} = 
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
(e^{i(\vec{k} \cdot \vec{R_{nm}}-\omega t)})(e^{i\vec{k} \cdot \vec{d_j}} - 1)
\tag{4}
$$

And Newton's 2nd Law
\begin{equation}
M \frac{\delta^2 \overrightarrow{U_{nm}}}{\delta t^2} = -M\omega^2 
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
(e^{i(\vec{k} \cdot \vec{R_{nm}}-\omega t)}) 
\tag{5}
\end{equation}

The time dependent term can be removed from both $Eq4$ and $Eq5$ as it is on both ends of the equation. Leaving the amplitudes for now.  
Going step by step through the rest of the equation
$$
(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} = 
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
[C_1(\hat{h_{j}})(e^{i\vec{k} \cdot \vec{h_j} } - 1) + C_2( \hat{p_{j}})(e^{i\vec{k} \cdot \vec{p_j} } - 1)]
\tag{6}
$$

Before building the final Dynamic Matrix the forces must be broken into the coordinate parts. This is done by taking the dot product of the vector displacement direction
$$
\begin{align}
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{x}) = [(\overrightarrow{U_{n+d_{j,x}}} - \overrightarrow{U_{n,x}}) \cdot \hat{d_{j,x}} + (\overrightarrow{U_{n+d_{j,y}}} - \overrightarrow{U_{n,y}}) \cdot \hat{d_{j,y}} ]\hat{d_{j,x}} \\
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{y}) = [(\overrightarrow{U_{n+d_{j,x}}} - \overrightarrow{U_{n,x}}) \cdot \hat{d_{j,x}} + (\overrightarrow{U_{n+d_{j,y}}} - \overrightarrow{U_{n,y}}) \cdot \hat{d_{j,y}} ]\hat{d_{j,y}}
\end{align}
$$

Applied to the general equation this becomes 
$$
\begin{align}
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{x}) = C_1 (\hat{h_{j,x}}^2 + \hat{h_{j,x}} \hat{h_{j,y}})(e^{i\vec{k} \cdot \vec{h_j} } - 1) + C_2 (\hat{p_{j,x}}^2 + \hat{p_{j,x}} \hat{p_{j,y}})(e^{i\vec{k} \cdot \vec{p_j} } - 1) \\
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{y}) =  C_1 (\hat{h_{j,y}}  \hat{h_{j,x}} + \hat{h_{j,y}}^2)(e^{i\vec{k} \cdot \vec{h_j} } - 1) + C_2 (\hat{p_{j,y}}  \hat{p_{j,x}} + \hat{p_{j,y}}^2)(e^{i\vec{k} \cdot \vec{p_j} } - 1) 
\end{align}
\tag{7}
$$


Solving through all nearest neighbors $j$ $Eq7$ produces
$$
\begin{align}
& \large \hat{x}-direction \\
& j = 1 \\
& C_1 (1^2 + (1)0)(e^{i\hat{k}a\hat{x}} -1) + C_2((\frac{1}{\sqrt{2}})^2 + \frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}})(e^{i\hat{k}(a\hat{x} +a\hat{y})} -1) \\
&= C_1(e^{i\hat{k_x}a} -1) + C_2(e^{i(\hat{k_x} + \hat{k_y})a} -1)  \\
& j =2 \\
& C_1 (0^2 + 0(1))(e^{i\hat{k}a\hat{y}} -1) + C_2((-\frac{1}{\sqrt{2}})^2 + -\frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}})(e^{i\hat{k}(-a\hat{x} +a\hat{y})} -1) \\
&= 0 \\
& j = 3 \\
& C_1 ((-1)^2 + (-1)0)(e^{i\hat{k}(-a\hat{x})} -1) + C_2((-\frac{1}{\sqrt{2}})^2 + (-\frac{1}{\sqrt{2}})(-\frac{1}{\sqrt{2}}))(e^{i\hat{k}(-a\hat{x} -a\hat{y})} -1) \\
&= C_1(e^{-i\hat{k_x}a} -1) + C_2(e^{-i(\hat{k_x} + \hat{k_y})a} -1) \\
& j = 4 \\
& C_1 (0^2 + 0(-1))(e^{i\hat{k}(-a\hat{y})} -1) + C_2((\frac{1}{\sqrt{2}})^2 + \frac{1}{\sqrt{2}} (-\frac{1}{\sqrt{2}}))(e^{i\hat{k}(a\hat{x} -a\hat{y})} -1) \\
&= 0 \\
& \text{Net X contribution} \\
&= C_1(e^{i\hat{k_x}a} -1) + C_2(e^{i(\hat{k_x} + \hat{k_y})a} -1) + C_1(e^{-i\hat{k_x}a} -1) + C_2(e^{-i(\hat{k_x} + \hat{k_y})a} -1) \\
\\
\\
& \large \hat{y}-direction \\
& j = 1 \\
& C_1 (0(1) + 0^2)(e^{i\hat{k}a\hat{x}} -1) + C_2(\frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}} + (\frac{1}{\sqrt{2}})^2 )(e^{i\hat{k}(a\hat{x} +a\hat{y})} -1) \\
&= C_2(e^{i(\hat{k_x} + \hat{k_y})a} -1) \\
& j =2 \\
& C_1 ((1)0 + 1^2)(e^{i\hat{k}a\hat{y}} -1) + C_2(-\frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}} + (\frac{1}{\sqrt{2}})^2 )(e^{i\hat{k}(-a\hat{x} +a\hat{y})} -1) \\
&= C_1(e^{i\hat{k_y}a} -1)  \\
& j = 3 \\
& C_1 (0(-1) + 0^2)(e^{i\hat{k}(-a\hat{x})} -1) + C_2((-\frac{1}{\sqrt{2}} )(-\frac{1}{\sqrt{2}}) + (-\frac{1}{\sqrt{2}})^2 )(e^{i\hat{k}(-a\hat{x} -a\hat{y})} -1) \\
&= C_2(e^{-i(\hat{k_x} + \hat{k_y})a} -1) \\
& j = 4 \\
& C_1 ((-1)0 + (-1)^2)(e^{i\hat{k}(-a\hat{y})} -1) + C_2(\frac{1}{\sqrt{2}} (-\frac{1}{\sqrt{2}}) + (-\frac{1}{\sqrt{2}})^2 )(e^{i\hat{k}(a\hat{x} -a\hat{y})} -1) \\
&= C_1(e^{-i\hat{k_y}a} -1) \\
& \text{Net Y contribution} \\
&= C_2(e^{i(\hat{k_x} + \hat{k_y})a} -1) + C_1(e^{i\hat{k_y}a} -1) + C_2(e^{-i(\hat{k_x} + \hat{k_y})a} -1) + C_1(e^{-i\hat{k_y}a} -1)
\end{align}
$$

Separating the parts based on the coordinates a Dynamic matrix of the form
$$
-M\omega^2
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
=
D_k
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
=
\begin{bmatrix}
D_{xx} & D_{xy} \\
D_{yx} & D_{yy}
\end{bmatrix}
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
$$

Is Built

$$
-
\begin{bmatrix}
M & 0 \\
0 & M
\end{bmatrix}
\omega^2
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
= 
\begin{bmatrix}
C_1(e^{i\hat{k_x}a}+e^{-i\hat{k_x}a}-2) + C_2(e^{i(\hat{k_x}+\hat{k_y})a}+e^{-i(\hat{k_x}+\hat{k_y})a}-2)  & C_2(e^{i(\hat{k_x}+\hat{k_y})a}+e^{-i(\hat{k_x}+\hat{k_y})a}-2) \\
C_2(e^{i(\hat{k_x}+\hat{k_y})a} + e^{-i(\hat{k_x}+\hat{k_y})a}-2) & C_1(e^{i\hat{k_y}a}+e^{-i\hat{k_y}a}-2) + C_2(e^{i(\hat{k_x}+\hat{k_y})a}+e^{-i(\hat{k_x}+\hat{k_y})a}-2)
\end{bmatrix}
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
\tag{8}
$$

 In this case the term $e^{ika} + e^{-ika}$ can be simplified first by using Euler's Identity for
 \begin{equation}
 e^{ika} + e^{-ika} = 2cos(ka)
 \end{equation}

and the matrix equation now comes out to be:
$$
-
\begin{bmatrix}
M & 0 \\
0 & M
\end{bmatrix}
\omega^2
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
= 
\begin{bmatrix}
2C_1[cos(\hat{k_x}a)-1] + 2C_2[cos((\hat{k_x}+\hat{k_y})a) -1]  & 2C_2[cos((\hat{k_x}+\hat{k_y})a) -1] \\
2C_2[cos((\hat{k_x}+\hat{k_y})a) -1] & 2C_1[cos(\hat{k_y}a) -1] + 2C_2[cos((\hat{k_x}+\hat{k_y})a) -1]
\end{bmatrix}
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
\tag{9}
$$

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt

# Physical parameters
m = 1.0    # mass
C1 = 2.0   # nearest-neighbor spring
C2 = 1.0   # 2nd nearest-neighbor spring
a = 1.0    # unit cell spacing

Nk = 300  # k-points per segment


# High-symmetry points
G = np.array([0.0, 0.0]) # Gamma points Γ where atoms move in phase 
X = np.array([np.pi/a, 0.0]) # X point where adjacent unit cells differ in phase. 
M = np.array([np.pi/a, np.pi/a]) # M points where there are maximum oscillations along x and y

# Build k-path: Γ → M → X → Γ
k_path = np.vstack([
    np.linspace(G, M, Nk, endpoint=False),
    np.linspace(M, X, Nk, endpoint=False),
    np.linspace(X, G, Nk),
])

# Distance along path
k_dist = np.zeros(len(k_path))
for i in range(1, len(k_path)):
    dk = k_path[i] - k_path[i-1]
    k_dist[i] = k_dist[i-1] + np.linalg.norm(dk)

# Eigenvalue calculation
omega = np.zeros((len(k_path), 2))

for i, (kx, ky) in enumerate(k_path):

    cx  = np.cos(kx * a)
    cy  = np.cos(ky * a)
    cxy = np.cos((kx + ky) * a)

    Dxx = 2*C1*(cx - 1) + C2*(cxy - 1)
    Dyy = 2*C1*(cy - 1) + C2*(cxy - 1)
    Dxy = C2*(cxy - 1)

    D = np.array([[Dxx, Dxy],
                  [Dxy, Dyy]], dtype=float)

    w2 = la.eigvalsh(-D / m)  # pull in negative from omega term in matrix equation before cancelling <0 values
    w2[w2 < 0] = 0.0          # numerical cleanup
    omega[i] = np.sqrt(w2)


# Plot dispersion
plt.figure(figsize=(7,4))

plt.plot(k_dist, omega[:,0], label='Acoustic')
plt.plot(k_dist, omega[:,1], label='Optical')

# High-symmetry ticks
xticks = [0, k_dist[Nk], k_dist[2*Nk], k_dist[-1]]
xtick_labels = [r'$\Gamma$', r'$X$', r'$M$', r'$\Gamma$']

plt.xticks(xticks, xtick_labels)

for x in xticks:
    plt.axvline(x, color='k', lw=0.5, alpha=0.5)

plt.xlabel('Wave vector')
plt.ylabel(r'$\omega$')
plt.legend()
plt.tight_layout()
plt.show()
```
