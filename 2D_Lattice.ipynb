## Standard Derivation of 2D Lattice

### Monoatomic Basis

For a square lattice with atoms of mass $M$ there will be phonons in both the horizontal and vertical axis for each atom in the crystal. As such the plane waves describing the system must be contained within a matrix of the form
$$
\overrightarrow{U_n} =
\begin{bmatrix} 
U_{nm,x} (t) \\
U_{nm,y} (t)
\end{bmatrix}
=
\large
\begin{bmatrix}
A_x \\
A_y 
\end{bmatrix}
e^{i(\vec{k} \cdot \vec{R_{nm}} -\omega t)}
\tag{1}
$$

in this case the vector, 
$$
\vec{k} =
\begin{bmatrix}
k_x \\
ky
\end{bmatrix}
$$
contains the allowed wave vectors, and the vector,
$$
\overrightarrow{R_{nm}} =
\begin{bmatrix}
n \cdot a_x \\
m \cdot a_y 
\end{bmatrix}
$$
contains the unit cell spacing of atoms within the crystal.

The displacement between any two points on the lattice can be described by the vector
\begin{equation}
\vec{h} = \overrightarrow{U_{n',m'}} - \overrightarrow{U_{n,m}}
\end{equation}

which has a direction of
\begin{equation}
\hat{h} = \frac{\vec{h}}{||\vec{h}||}
\end{equation}

The displacements in a square lattice depend on the position of the nearest neighbors. The first nearest neighbors will exist on the horizontal and vertical planes such that 
$$
\begin{aligned}
h_1 &= a\hat{x} \quad
h_2 = a\hat{y} \\
h_3 &= -a\hat{x} \quad
h_4 = -a\hat{y}
\end{aligned}
$$
While the second nearest neighbors exist along the diagonals with displacements
$$
\begin{aligned}
p_1 &= a\hat{x} + a\hat{y} \quad
p_2 = -a\hat{x} + a\hat{y} \\
p_3 &= -a\hat{x} - a\hat{y} \quad
p_4 = a\hat{x} - a\hat{y}
\end{aligned}
$$

The directions corresponding to each of these vectors can be solved out such that  
$$
\begin{aligned}
\hat{h_1} &= \hat{x} \quad
\hat{h_2} = \hat{y} \\
\hat{h_3} &= -\hat{x} \quad
\hat{h_4} = -\hat{y}
\end{aligned}
$$
$$
\begin{aligned}
\hat{p_1} &= \frac{1}{\sqrt{2}}(\hat{x} + \hat{y}) \quad
\hat{p_2} = \frac{1}{\sqrt{2}}(-\hat{x} + \hat{y}) \\
\hat{p_3} &= \frac{1}{\sqrt{2}}(-\hat{x} - \hat{y})\quad
\hat{p_4} = \frac{1}{\sqrt{2}}(\hat{x} - \hat{y})
\end{aligned}
$$

Note: $h_j$ and $p_j$ represent the same variable and are just different for ease of seeing which belongs to 1st or 2nd nearest neighbors

Similar to the systems for 1D photons the forces can be solved with Newton's 2nd Law in the general form

\begin{equation}
M \frac{\delta^2 \overrightarrow{U_{nm}}}{\delta t^2} = C_1\sum_{j=1,2,3,4} [(\overrightarrow{U_{n',m'}} - \overrightarrow{U_{n,m}}) \cdot \hat{h_j}]\hat{h_j} + C_2\sum_{j=1,2,3,4} [(\overrightarrow{U_{n',m'}} - \overrightarrow{U_{n,m}}) \cdot \hat{p_j}]\hat{p_j}  + C_3....
\tag{2}
\end{equation}
Where the equation can be expanded to include as many nearest neighbors as required. In this square lattice only up to 2nd nearest is required.    

Centering on site $(n,m)$ this form can be altered to 
\begin{equation}
M \frac{\delta^2 \overrightarrow{U_{nm}}}{\delta t^2} = C_1\sum_{j=1,2,3,4} [(\overrightarrow{U_{n+h_j}} - \overrightarrow{U_{n}}) \cdot \hat{h_j}]\hat{h_j} + C_2\sum_{j=1,2,3,4} [(\overrightarrow{U_{n+p_j}} - \overrightarrow{U_{n}}) \cdot \hat{p_j}]\hat{p_j} 
\tag{3}
\end{equation}

And the plane wave solutions from $Eq1$ can be called into the problem using the solution that
$$
\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}} = 
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
(e^{i(\vec{k} \cdot \vec{R_{nm}}-\omega t)})(e^{i\vec{k} \cdot \vec{d_j}} - 1)
\tag{4}
$$

And Newton's 2nd Law
\begin{equation}
M \frac{\delta^2 \overrightarrow{U_{nm}}}{\delta t^2} = -M\omega^2 
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
(e^{i(\vec{k} \cdot \vec{R_{nm}}-\omega t)}) 
\tag{5}
\end{equation}

The time dependent term can be removed from both $Eq4$ and $Eq5$ as it is on both ends of the equation. Leaving the amplitudes for now.  
Going step by step through the rest of the equation
$$
(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} = 
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
[C_1(\hat{h_{j}})(e^{i\vec{k} \cdot \vec{h_j} } - 1) + C_2( \hat{p_{j}})(e^{i\vec{k} \cdot \vec{p_j} } - 1)]
\tag{6}
$$

Before building the final Dynamic Matrix the forces must be broken into the coordinate parts. This is done by taking the dot product of the vector displacement direction
$$
\begin{align}
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{x}) = [(\overrightarrow{U_{n+d_{j,x}}} - \overrightarrow{U_{n,x}}) \cdot \hat{d_{j,x}} + (\overrightarrow{U_{n+d_{j,y}}} - \overrightarrow{U_{n,y}}) \cdot \hat{d_{j,y}} ]\hat{d_{j,x}} \\
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{y}) = [(\overrightarrow{U_{n+d_{j,x}}} - \overrightarrow{U_{n,x}}) \cdot \hat{d_{j,x}} + (\overrightarrow{U_{n+d_{j,y}}} - \overrightarrow{U_{n,y}}) \cdot \hat{d_{j,y}} ]\hat{d_{j,y}}
\end{align}
$$

Applied to the general equation this becomes  
Note: At this point in the derivation all values are scalar. The hat remains merely to denote the described term
$$
\begin{align}
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{x}) = C_1 (\hat{h_{j,x}}^2 + \hat{h_{j,x}} \hat{h_{j,y}})(e^{i\vec{k} \cdot \vec{h_j} } - 1) + C_2 (\hat{p_{j,x}}^2 + \hat{p_{j,x}} \hat{p_{j,y}})(e^{i\vec{k} \cdot \vec{p_j} } - 1) \\
[(\overrightarrow{U_{n+d_j}} - \overrightarrow{U_{n}}) \cdot \hat{d_j} ](\hat{d_j} \cdot \hat{y}) =  C_1 (\hat{h_{j,y}}  \hat{h_{j,x}} + \hat{h_{j,y}}^2)(e^{i\vec{k} \cdot \vec{h_j} } - 1) + C_2 (\hat{p_{j,y}}  \hat{p_{j,x}} + \hat{p_{j,y}}^2)(e^{i\vec{k} \cdot \vec{p_j} } - 1) 
\end{align}
\tag{7}
$$


Solving through all nearest neighbors $j$ $Eq7$ produces
$$
\begin{flalign}
\large \hat{x}-direction &&\\
j = 1 &&\\
& C_1 (1^2 + (1)0)(e^{i\vec{k}a\hat{x}} -1) + C_2((\frac{1}{\sqrt{2}})^2 + \frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}})(e^{i\vec{k}(a\hat{x} +a\hat{y})} -1) \\
&= C_1(e^{-ik_x a} -1) + C_2(e^{i(k_x + k_y)a} -1)  \\
j = 2 &&\\
& C_1 (0^2 + 0(1))(e^{i\vec{k}a\hat{y}} -1) + C_2((-\frac{1}{\sqrt{2}})^2 + -\frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}})(e^{i\vec{k}(-a\hat{x} +a\hat{y})} -1) \\
&= 0 \\
j = 3 &&\\
& C_1 ((-1)^2 + (-1)0)(e^{i\vec{k}(-a\hat{x})} -1) + C_2((-\frac{1}{\sqrt{2}})^2 + (-\frac{1}{\sqrt{2}})(-\frac{1}{\sqrt{2}}))(e^{i\vec{k}(-a\hat{x} -a\hat{y})} -1) \\
&= C_1(e^{-ik_x a} -1) + C_2(e^{-i(k_x + k_y)a} -1) \\
j = 4 &&\\
& C_1 (0^2 + 0(-1))(e^{i\vec{k}(-a\hat{y})} -1) + C_2((\frac{1}{\sqrt{2}})^2 + \frac{1}{\sqrt{2}} (-\frac{1}{\sqrt{2}}))(e^{i\vec{k}(a\hat{x} -a\hat{y})} -1) \\
&= 0 \\
\text{Net X contribution} &&\\
&= C_1(e^{-ik_x a} -1) + C_2(e^{i(k_x + k_y)a} -1) + C_1(e^{-ik_x a} -1) + C_2(e^{-i(k_x + k_y)a} -1) \\
\\
\\
\large \hat{y}-direction &&\\
j = 1 &&\\
& C_1 (0(1) + 0^2)(e^{i\vec{k}a\hat{x}} -1) + C_2(\frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}} + (\frac{1}{\sqrt{2}})^2 )(e^{i\vec{k}(a\hat{x} +a\hat{y})} -1) \\
&= C_2(e^{i(k_x + k_y)a} -1) \\
j = 2 &&\\
& C_1 ((1)0 + 1^2)(e^{i\vec{k}a\hat{y}} -1) + C_2(-\frac{1}{\sqrt{2}} \frac{1}{\sqrt{2}} + (\frac{1}{\sqrt{2}})^2 )(e^{i\vec{k}(-a\hat{x} +a\hat{y})} -1) \\
&= C_1(e^{ik_y a} -1)  \\
j = 3 &&\\
& C_1 (0(-1) + 0^2)(e^{i\vec{k}(-a\hat{x})} -1) + C_2((-\frac{1}{\sqrt{2}} )(-\frac{1}{\sqrt{2}}) + (-\frac{1}{\sqrt{2}})^2 )(e^{i\vec{k}(-a\hat{x} -a\hat{y})} -1) \\
&= C_2(e^{-i(k_x + k_y)a} -1) \\
j = 4 &&\\
& C_1 ((-1)0 + (-1)^2)(e^{i\vec{k}(-a\hat{y})} -1) + C_2(\frac{1}{\sqrt{2}} (-\frac{1}{\sqrt{2}}) + (-\frac{1}{\sqrt{2}})^2 )(e^{i\vec{k}(a\hat{x} -a\hat{y})} -1) \\
&= C_1(e^{-ik_y a} -1) \\
\text{Net Y contribution} &&\\
&= C_2(e^{i(k_x + k_y)a} -1) + C_1(e^{ik_y a} -1) + C_2(e^{-i(k_x + k_y)a} -1) + C_1(e^{-ik_y a} -1)
\end{flalign}
$$

Separating the parts based on the coordinates a Dynamic matrix of the form
$$
-M\omega^2
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
=
D_k
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
=
\begin{bmatrix}
D_{xx} & D_{xy} \\
D_{yx} & D_{yy}
\end{bmatrix}
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
$$

Is Built

$$
-
\begin{bmatrix}
M & 0 \\
0 & M
\end{bmatrix}
\omega^2
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
= 
\begin{bmatrix}
C_1(e^{-ik_x a}+e^{-ik_x a}-2) + C_2(e^{i(k_x +k_y)a}+e^{-i(k_x +k_y)a}-2)  & C_2(e^{i(k_x +k_y)a}+e^{-i(k_x +k_y)a}-2) \\
C_2(e^{i(k_x +k_y)a} + e^{-i(k_x +k_y)a}-2) & C_1(e^{ik_y a}+e^{-ik_y a}-2) + C_2(e^{i(k_x +k_y)a}+e^{-i(k_x +k_y)a}-2)
\end{bmatrix}
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
\tag{8}
$$

In this case the term $e^{ika} + e^{-ika}$ can be simplified first by using Euler's Identity for
\begin{equation}
e^{ika} + e^{-ika} = 2cos(ka)
\end{equation}

and the matrix equation now comes out to be:
$$
-
\begin{bmatrix}
M & 0 \\
0 & M
\end{bmatrix}
\omega^2
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
= 
\begin{bmatrix}
2C_1[cos(k_x a)-1] + 2C_2[cos((k_x +k_y)a) -1]  & 2C_2[cos((k_x +k_y)a) -1] \\
2C_2[cos((k_x +k_y)a) -1] & 2C_1[cos(k_ya) -1] + 2C_2[cos((k_x +k_y)a) -1]
\end{bmatrix}
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
\tag{9}
$$

However, the diagonal terms related to $C_2$ have overlapping force vectors when summed individually. Such as both the $1st$ and $4th$ quadrants using $+\hat{x}$ vectors. This is fixed by dividing the $C_2$ sums by $2$ to avoid double counting. The resulting matrix now becomes

$$
-
\begin{bmatrix}
M & 0 \\
0 & M
\end{bmatrix}
\omega^2
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
= 
\begin{bmatrix}
2C_1[cos(k_x a)-1] + C_2[cos(k_x +k_y)a) -1]  & C_2[cos((k_x +k_y)a) -1] \\
C_2[cos((k_x +k_y)a) -1] & 2C_1[cos(k_y a) -1] + C_2[cos((k_x +k_y)a) -1]
\end{bmatrix}
\begin{bmatrix}
A_x \\
A_y
\end{bmatrix}
\tag{10}
$$
Which is the final solution.

### Diatomic (Honeycomb) Lattice

Fig 1: Display of a diatomic honeycomb lattice. Blue points represent atom A in these calculations. Specifically the right most atom A in each hexagonal array

#### Description of the Honeycomb Lattice
The honeycomb lattice has two standard basis vectors stemming from the left most atom in each hexagon. These vectors are denoted as
$$
\vec{a_1} = a \begin{pmatrix} \frac{3}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix} \quad 
\vec{a_2} = a \begin{pmatrix} \frac{3}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix}
$$

It should be noted that the value $a$ for unit cell spacing is not the same as the vectors  
The honeycomb lattice has bond displacements similar to that of a triangle lattice (the system can be seen as an overlap of 2 triangular lattice, but that is not covered here). Describing the displacements in terms of the basis vectors can be done as such    
The $1st$ nearest-neighbors vectors are all $A$-$B$ bonds between alternating atom types
$$
\begin{align}
\vec{h_1} = a \begin{pmatrix} 1 \\ 0 \end{pmatrix} \quad
\vec{h_2} = \vec{h_1} - \vec{a_1} = a \begin{pmatrix} -\frac{1}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix} \quad
\vec{h_3} = \vec{h_1} - \vec{a_2} = a \begin{pmatrix} -\frac{1}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix} \quad
\end{align}
$$
All of which simply have directions 
$$
\hat{h_j} = \frac{\vec{h_j}}{a}
$$
 
The $2nd$ nearest-neighbor are all like atom $A$-$A$ bonds. There are $6$ of these bonds and they are the same as a standard triangular lattice
$$
\begin{align}
\vec{p}_{1,3} = \pm \vec{a_1} = \pm a \begin{pmatrix} \frac{3}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix} \quad
\vec{p}_{2,3} = \pm \vec{a_2} = \pm a \begin{pmatrix} \frac{3}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix} \quad
\vec{p}_{5,6} = \pm (\vec{a_1} - \vec{a_2}) = \pm a \begin{pmatrix} 0 \\ \sqrt{3} \end{pmatrix} 
\end{align}
$$
With the corresponding directions
$$
\begin{align}
\hat{p}_{1,3} = \pm \begin{pmatrix} \frac{\sqrt{3}}{2} \\ \frac{1}{2} \end{pmatrix} \quad
\hat{p}_{2,3} = \pm \begin{pmatrix} \frac{\sqrt{3}}{2} \\ -\frac{1}{2} \end{pmatrix} \quad
\hat{p}_{5,6} = \pm \begin{pmatrix} 0 \\ 1 \end{pmatrix} 
\end{align}
$$

Finally, there are 3 $3rd$ nearest-neighbors which exist in opposite positions to the $1st$ nearest-neighbors as more $A$-$B$ bonds, however these have lengths of $2a$ now and reach across the hexagonal plane in opposite directions from $\vec{h_j}$
$$
\begin{align}
\vec{b_1} = -2\hat{h_1} \quad
\vec{b_2} = \vec{h_2} + \vec{a_2} = a \begin{pmatrix} 1 \\ -\sqrt{3} \end{pmatrix} \quad
\vec{b_3} = \vec{h_3} + \vec{a_1} = a \begin{pmatrix} 1 \\ \sqrt{3} \end{pmatrix} \quad
\end{align}
$$
With the corresponding directions
$$
\begin{align}
\hat{b_1} = \begin{pmatrix} -1 \\ 0 \end{pmatrix} \quad
\hat{b_2} = \begin{pmatrix} \frac{1}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix} \quad
\hat{b_3} = \begin{pmatrix} \frac{1}{2} \\ \frac{\sqrt{3}}{2} \end{pmatrix}
\end{align}
$$

#### Hooke's Law with Projection 
A spring connecting $R$ to $R + \delta$ represents a change in length along the $\hat{e}$ direction
$$
\begin{equation} \Delta \ell (\vec{R},\vec{\delta}) = \hat{e} \cdot [\vec{U}(\vec{R} + \vec{\delta}) - \vec{U}] \end{equation}
\tag{1}
$$

where the force causing this displacement is represented by some spring constant, $C$, along a direction, $\hat{e}$
$$
\begin{equation} \vec{F} (\vec{R},\vec{\delta}) = C \Delta \ell (\vec{R},\vec{\delta})\hat{e} = C [\hat{e} \cdot [\vec{U}(\vec{R} + \vec{\delta}) - \vec{U}]]\hat{e} \end{equation}
\tag{2}
$$

Using the vector relationship that
$$
\begin{equation} \vec{A}(\vec{A} \cdot \vec{B}) = (\vec{A} \vec{A}^T)\vec{B} = (\vec{A} \otimes \vec{A})\vec{B} \end{equation}
$$

the equation can be rearrange in terms of a projection on the \hat{e} direction. Note that multiplying a vector by it's transpose is the same as the outer product notation. The force is now 
$$
\begin{equation} \vec{F} (\vec{R},\vec{\delta}) = C(\hat{e} \otimes \hat{e})[\vec{U}(\vec{R} + \vec{\delta}) - \vec{U}] \end{equation}
\tag{3}
$$

Now using Newton's $2nd$ Law the force on atom $A$ due to a displacement of atom $B$ is represented by 
$$
\begin{equation} 
M \ddot{U_A}(\vec{R}) = 
C_1 \sum_{j=1}^{3} (\hat{h_j} \otimes \hat{h_j}) [\vec{U_B}(\vec{R} + \vec{h_j}) - \vec{U_A}] +
C_3 \sum_{j=1}^{3} (\hat{b_j} \otimes \hat{b_j}) [\vec{U_B}(\vec{R} + \vec{b_j}) - \vec{U_A}]
\end{equation}
\tag{4}
$$

And the forces on like atoms $A$-$A$ is seen as 
$$
\begin{equation} 
M \ddot{U_A}(\vec{R}) = 
C_2 \sum_{j=1}^{6} (\hat{p_j} \otimes \hat{p_j}) [\vec{U_A}(\vec{R} + \vec{p_j}) - \vec{U_A}]
\end{equation}
\tag{5}
$$

#### Projectors
Before making trying for a solution to the system the projectors should be knows to simplify calculations
$$
\begin{align}
&(\hat{h_1} \otimes \hat{h_1}) = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} \quad
(\hat{h_2} \otimes \hat{h_2}) = \begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} \quad
(\hat{h_3} \otimes \hat{h_3}) = \begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} \\ 
\\
&(\hat{p}_{1,3} \otimes \hat{p}_{1,3}) = \begin{pmatrix} \frac{3}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} \quad
(\hat{p}_{2,4} \otimes \hat{p}_{2,4}) = \begin{pmatrix} \frac{3}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} \quad
(\hat{p}_{5,6} \otimes \hat{p}_{5,6}) = \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} \\
\\
&(\hat{b_1} \otimes \hat{b_1}) = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} \quad
(\hat{b_2} \otimes \hat{b_2}) = \begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} \quad
(\hat{b_3} \otimes \hat{b_3}) = \begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} 
\end{align}
$$

#### Bloch's Theorem 
Applying Bloch's theorem says that any wavefunction in a periodic crystal can be described as a moving wave of the form 
$$
\vec{U}(\vec{R},t) = \vec{U}e^{i(\vec{k} \cdot \vec{R} -\omega t)}
$$

Applying this to the atoms in this system yields
$$
\vec{U_A}(\vec{R},t) = \vec{U_A}e^{i(\vec{k} \cdot \vec{R} -\omega t)} \quad
\vec{U_B}(\vec{R},t) = \vec{U_B}e^{i(\vec{k} \cdot \vec{R} -\omega t)} \quad
\vec{U_B}(\vec{R} + \vec{\delta},t) = \vec{U_B}(\vec{R},t) e^{i(\vec{k} \cdot \vec{\delta})} 
$$

Which when plugged into $Eq4$ and $Eq5$ yield
$$
\begin{align} 
-M \omega^2 \vec{U_A} = 
C_1 \sum_{j=1}^{3} (\hat{h_j} \otimes \hat{h_j}) [\vec{U_B} e^{i(\vec{k} \cdot \vec{h_j})} - \vec{U_A}] +
C_3 \sum_{j=1}^{3} (\hat{b_j} \otimes \hat{b_j}) [\vec{U_B} e^{i(\vec{k} \cdot \vec{b_j})} - \vec{U_A}] \\
M \omega^2 \vec{U_A} = 
C_1 \sum_{j=1}^{3} (\hat{h_j} \otimes \hat{h_j}) [\vec{U_A} - \vec{U_B} e^{i(\vec{k} \cdot \vec{h_j})}] +
C_3 \sum_{j=1}^{3} (\hat{b_j} \otimes \hat{b_j}) [\vec{U_A} - \vec{U_B} e^{i(\vec{k} \cdot \vec{b_j})}]
\end{align}
\tag{6}
$$
$$
\begin{align} 
-M \omega^2 \vec{U_A} = 
C_2 \sum_{j=1}^{6} (\hat{p_j} \otimes \hat{p_j}) [\vec{U_A}e^{i(\vec{k} \cdot \vec{p_j})} - \vec{U_A}] \\
M \omega^2 \vec{U_A} = 
C_2 \sum_{j=1}^{6} (\hat{p_j} \otimes \hat{p_j}) [\vec{U_A} - \vec{U_A}e^{i(\vec{k} \cdot \vec{p_j})}] 
\end{align}
\tag{7}
$$

At this point it can already be clearly seen how there is a restoring $A$-$A$ interaction even within the $A$-$B$ force term. This term can be defined as
$$
\begin{align}
\sum_{j=1}^3 P_j &= C_1 \sum_{j=1}^3(\hat{h_j}\hat{h_j}^T) + C_3 \sum_{j=1}^3(\hat{b_j}\hat{b_j}^T) \\
\sum_{j=1}^3 P_j &= 
\begin{pmatrix} 1 + \frac{1}{4} + \frac{1}{4} & 0 \\ 0 & 0 + \frac{3}{4} + \frac{3}{4} \end{pmatrix} = 
\frac{3}{2} \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} = \frac{3}{2}I(C_1 + C_3)
\end{align}
$$

#### Dynamic Matrices
The net forces for both the $A$-$A$ and $A$-$B$ interactions can now be fully described as
$$
A \text{-} A = \frac{3}{2}I(C_1 + C_3) + 
C_2 \left[\begin{pmatrix} \frac{3}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} (1 - e^{i(\vec{k} \cdot \vec{p}_{1,3})}) +
\begin{pmatrix} \frac{3}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{1}{4} \end{pmatrix} (1 - e^{i(\vec{k} \cdot \vec{p}_{2,4})}) +
\begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} (1 - e^{i(\vec{k} \cdot \vec{p}_{5,6})}) \right]
$$
$$
A \text{-} B = 
-C_1 \left[\begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} e^{i(\vec{k} \cdot \vec{h_1})} + 
\begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{h_2})} +
\begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{h_3})} \right]
-C_3 \left[\begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} e^{i(\vec{k} \cdot \vec{b_1})} + 
\begin{pmatrix} \frac{1}{4} & -\frac{\sqrt{3}}{4} \\ -\frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{b_2})} +
\begin{pmatrix} \frac{1}{4} & \frac{\sqrt{3}}{4} \\ \frac{\sqrt{3}}{4} & \frac{3}{4} \end{pmatrix} e^{i(\vec{k} \cdot \vec{b_3})} \right]
$$

Note how the leading terms in the $A$-$B$ interactions are all negative due to the swap around of the force equation  
The matrices describing all these forces take the form
$$
D^{AA} = \begin{bmatrix} D^{AA}_{xx} & D^{AA}_{xy} \\ D^{AA}_{yx} & D^{AA}_{yy} \end{bmatrix} \quad
\begin{align}
& D^{AA}_{xx} = \quad
\frac{3}{2}(C_1 + C_3) + C_2 [ \frac{3}{4}(2 - e^{i(\vec{k} \cdot \vec{p}_{1,3})} - e^{i(\vec{k} \cdot \vec{p}_{2,4})}) + (1 - e^{i(\vec{k} \cdot \vec{p}_{5,6})}) ] \\
& D^{AA}_{xy} \text{ and } D^{AA}_{yx} = \quad
C_2 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{p}_{2,4})} - e^{i(\vec{k} \cdot \vec{p}_{1,3})}) ] \\
& D^{AA}_{yy} = \quad
\frac{3}{2}(C_1 + C_3) + C_2 [ \frac{1}{4}(2 - e^{i(\vec{k} \cdot \vec{p}_{1,3})} - e^{i(\vec{k} \cdot \vec{p}_{2,4})}) ] \\
\end{align}
$$

$$
D^{AB} = \begin{bmatrix} D^{AB}_{xx} & D^{AB}_{xy} \\ D^{AB}_{yx} & D^{AB}_{yy} \end{bmatrix} \quad
\begin{align}
& D^{AB}_{xx} = \quad
-C_1 [ e^{i(\vec{k} \cdot \vec{h_1})} + \frac{1}{4}(e^{i(\vec{k} \cdot \vec{h_2})} + e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ e^{i(\vec{k} \cdot \vec{b_1})} + \frac{1}{4}(e^{i(\vec{k} \cdot \vec{b_2})} + e^{i(\vec{k} \cdot \vec{b_3})}) ] \\
& D^{AB}_{xy} \text{ and } D^{AB}_{yx} = \quad
-C_1 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{h_2})} - e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ \frac{\sqrt{3}}{4}(e^{i(\vec{k} \cdot \vec{b_3})} - e^{i(\vec{k} \cdot \vec{b_2})}) ] \\
& D^{AB}_{yy} = \quad
-C_1 [ \frac{3}{4}(e^{i(\vec{k} \cdot \vec{h_2})} + e^{i(\vec{k} \cdot \vec{h_3})}) ]
-C_3 [ \frac{3}{4}(e^{i(\vec{k} \cdot \vec{b_2})} + e^{i(\vec{k} \cdot \vec{b_3})}) ]
\end{align}
$$

**Please note that each p term represents $2$ vectors in the $\pm$ domain**

#### Full Dynamic Matrix

The Dynamic matrix is now defined as
$$
D(k) =
\begin{bmatrix}
D^{AA} & D^{AB} \\
D^{BA} & D^{BB}
\end{bmatrix}
=
\begin{bmatrix}
D^{AA}_{xx} & D^{AA}_{xy} & D^{AB}_{xx} & D^{AB}_{xy} \\
D^{AA}_{yx} & D^{AA}_{yy} & D^{AB}_{yx} & D^{AB}_{yy} \\
D^{BA}_{xx} & D^{BA}_{xy} & D^{BB}_{xx} & D^{BB}_{xy} \\
D^{BA}_{yx} & D^{BA}_{yy} & D^{BB}_{yx} & D^{BB}_{yy} 
\end{bmatrix}
$$

However the only values calculated were $D^{AB}$ and $D^{AA}$. This is fine due to the Hermitian symmetry of the Dynamic matrix. The term $D^{BB} = D^{AA}$ and the term $D^{BA} = D^{AB\dagger}$. Therefor the matrix can be decribed as 
$$
D(k) =
\begin{bmatrix}
D^{AA} & D^{AB} \\
D^{AB\dagger} & D^{AA}
\end{bmatrix}
$$

and used to solve
$$
\omega^2
\begin{bmatrix}
M_A & 0 & 0 & 0 \\
0 & M_A & 0 & 0 \\
0 & 0 & M_B & 0 \\
0 & 0 & 0 & M_B 
\end{bmatrix}
U_A
=
\begin{bmatrix}
D^{AA} & D^{AB} \\
D^{AB\dagger} & D^{AA}
\end{bmatrix}
U_A
\tag{8}
$$

Which is now the form of a standard Eigenvalue problem and can be evaluated with linear algebra solvers

#### High Symmetry Points
The high symmetry points represent the actions of a lattice at specific points. In the case of a hexagonal lattice there are three key symmetry points defined as $\Gamma, K, M$

These points define specific locations along the lattice within the Brillouin Zone, and as such are described by the reciprocal lattice vectors
$$
\begin{align}
b_1 = \frac{2\pi}{a} \begin{pmatrix} \frac{1}{3} \\ \frac{1}{\sqrt{3}} \end{pmatrix} \quad
b_2 = \frac{2\pi}{a} \begin{pmatrix} \frac{1}{3} \\ -\frac{1}{\sqrt{3}} \end{pmatrix}
\end{align}
$$

The $\Gamma$ point represents the origin of the BZ and as such sits as 
$$ \Gamma = \begin{pmatrix} 0 \\ 0 \end{pmatrix} $$
The $K$ point represents any particular corner of the BZ and as such there are 6 different variations. This $K$ is defined as 
$$ 
K = \frac{1}{3}(b_1,b_2) = \frac{2\pi}{3a} \begin{pmatrix} \frac{1}{3} \\ \frac{1}{3\sqrt{3}} \end{pmatrix}
$$
The $M$ point represents any particular midpoint of the BZ and there are also 6 different variations. This $M$ is defined as 
$$ 
M = \frac{1}{2}(b_1 + b_2) = \frac{2\pi}{a} \begin{pmatrix} \frac{1}{3} \\ 0 \end{pmatrix}
$$

### Monoatomic Lattice Plot

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt

# Physical parameters
m = 1.0    # mass
C1 = 2.0   # nearest-neighbor spring
C2 = 1.0   # 2nd nearest-neighbor spring
a = 1.0    # unit cell spacing

Nk = 300  # k-points per segment


# High-symmetry points
G = np.array([0.0, 0.0]) # Gamma points Γ where atoms move in phase 
X = np.array([np.pi/a, 0.0]) # X point where adjacent unit cells differ in phase. 
M = np.array([np.pi/a, np.pi/a]) # M points where there are maximum oscillations along x and y

# Build k-path: Γ → M → X → Γ
k_path = np.vstack([
    np.linspace(G, M, Nk, endpoint=False),
    np.linspace(M, X, Nk, endpoint=False),
    np.linspace(X, G, Nk),
])

# Distance along path
k_dist = np.zeros(len(k_path))
for i in range(1, len(k_path)):
    dk = k_path[i] - k_path[i-1]
    k_dist[i] = k_dist[i-1] + np.linalg.norm(dk)

# Eigenvalue calculation
omega = np.zeros((len(k_path), 2))

for i, (kx, ky) in enumerate(k_path):

    cx  = np.cos(kx * a)
    cy  = np.cos(ky * a)
    cxy = np.cos((kx + ky) * a)

    Dxx = 2*C1*(cx - 1) + C2*(cxy - 1)
    Dyy = 2*C1*(cy - 1) + C2*(cxy - 1)
    Dxy = C2*(cxy - 1)

    D = np.array([[Dxx, Dxy],
                  [Dxy, Dyy]], dtype=float)

    w2 = la.eigvalsh(-D / m)  # pull in negative from omega term in matrix equation before cancelling <0 values
    w2[w2 < 0] = 0.0          # numerical cleanup
    omega[i] = np.sqrt(w2)


# Plot dispersion
plt.figure(figsize=(7,4))

plt.plot(k_dist, omega[:,0], label='Transverse Acoustic')
plt.plot(k_dist, omega[:,1], label='Longitudinal Acoustic')

# High-symmetry ticks
xticks = [0, k_dist[Nk], k_dist[2*Nk], k_dist[-1]]
xtick_labels = [r'$\Gamma$', r'$X$', r'$M$', r'$\Gamma$']

plt.xticks(xticks, xtick_labels)

for x in xticks:
    plt.axvline(x, color='k', lw=0.5, alpha=0.5)

plt.xlabel('Wave vector')
plt.ylabel(r'$\omega$')
plt.legend()
plt.tight_layout()
plt.show()
```

### Diatomic (Honeycomb) Lattice

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt

# Physical parameters
mA = 1.0 # mass of A
mB = 2.0 # mass of B
C1 = 5.0
C2 = 2.0
C3 = 1.0
a = 1.0 # unit cell spacing

Nk = 3000

# Mass matrix
mass = np.zeros((4, 4))
np.fill_diagonal(mass, [mA, mA, mB, mB])

# Displacement Vectors
a1 = a*np.array([3/2 , np.sqrt(3)/2])
a2 = a*np.array([3/2 , -np.sqrt(3)/2])

h1 = a*np.array([1 , 0])
h2 = h1 - a1
h3 = h1 - a2

p1 = a1
p2 = a2
p3 = -a1
p4 = -a2
p5 = (a1 - a2)
p6 = -(a1 - a2)

b1 = -2*h1
b2 = h2 + a2
b3 = h3 + a1

# High-symmetry points
G = np.array([0.0, 0.0])
M = np.array([2*np.pi/(3*a) , 0.0])  # midpoint of BZ edge 
K = np.array([2*np.pi/(3*a), 2*np.pi/(3*np.sqrt(3)*a)])  # corner of BZ

k_path = np.vstack([
    np.linspace(G, K, Nk, endpoint=False),
    np.linspace(K, M, Nk, endpoint=False),  
    np.linspace(M, G, Nk, endpoint=True) ])

k_dist = np.zeros(len(k_path))
for i in range(1, len(k_path)):
    dk = k_path[i] - k_path[i-1]
    k_dist[i] = k_dist[i-1] + np.linalg.norm(dk)

# Initialize eigenvalues
omega = np.zeros((len(k_path), 4))

# Looping through k-points to build force matrices
for i, (kx, ky) in enumerate(k_path):
    
    # Shorthand notation
    k = np.array([kx, ky])
    e = np.exp
    i_c = 1j
    s3 = np.sqrt(3)

    # Phase factors
    # =========================
    
    # h-vectors
    eh1 = e(i_c * np.dot(k, h1))
    eh2 = e(i_c * np.dot(k, h2))
    eh3 = e(i_c * np.dot(k, h3))

    # p-vectors 
    ep1 = e(i_c * np.dot(k, p1))
    ep3 = e(i_c * np.dot(k, p3))
    ep2 = e(i_c * np.dot(k, p2))
    ep4 = e(i_c * np.dot(k, p4))
    ep5 = e(i_c * np.dot(k, p5))
    ep6 = e(i_c * np.dot(k, p6))

    # b-vectors
    eb1 = e(i_c * np.dot(k, b1))
    eb2 = e(i_c * np.dot(k, b2))
    eb3 = e(i_c * np.dot(k, b3))

    # D_AA components
    D_AA_xx = (
        3/2*(C1 + C3) + C2 * (3/4*(4 - ep1 - ep3 - ep2 - ep4 ) + (2 - ep5 - ep6)) )

    D_AA_xy = C2 * (s3/4 * (ep2 + ep4 - ep1 - ep3))
    D_AA_yx = D_AA_xy

    D_AA_yy = (
        3/2*(C1 + C3) + C2 * ( 1/4*(4 - ep1 - ep3 - ep2 - ep4 )) )

    D_AA = np.array([
        [D_AA_xx, D_AA_xy],
        [D_AA_yx, D_AA_yy] ])

    # D_AB components
    D_AB_xx = (
        -C1*(eh1 + 1/4*(eh2 + eh3)) -C3*(eb1 + 1/4*(eb2 + eb3)) )

    D_AB_xy = (
        -C1*(s3/4*(eh2 - eh3)) -C3*(s3/4*(eb3 - eb2)) )

    D_AB_yx = D_AB_xy

    D_AB_yy = (
        -C1*(3/4*(eh2 + eh3)) -C3*(3/4*(eb2 + eb3)) )

    D_AB = np.array([
        [D_AB_xx, D_AB_xy],
        [D_AB_yx, D_AB_yy] ])

    # Hermitian conjugate
    D_BA = D_AB.conj().T

    # Same-sublattice B
    D_BB = D_AA.copy()

    # Full matrix
    D = np.block([
        [D_AA, D_AB],
        [D_BA, D_BB]
    ])

    w2 = la.eigvalsh(D, mass)
    w2[w2 < 0] = 0.0
    omega[i] = np.sqrt(w2)


# Plotting
plt.figure(figsize=(8,5))

for n in range(4):
    if n < 2:
        plt.plot(k_dist, omega[:,n], 'b', label='Acoustic' if n==0 else "")
    else:
        plt.plot(k_dist, omega[:,n], 'r', label='Optical' if n==2 else "")

xticks = [0, k_dist[Nk], k_dist[2*Nk], k_dist[-1]]
xtick_labels = [r'$\Gamma$', r'$K$', r'$M$', r'$\Gamma$']

plt.xticks(xticks, xtick_labels)
for x in xticks:
    plt.axvline(x, color='k', lw=0.5, alpha=0.3, linestyle='--')

plt.xlabel('Wave vector')
plt.ylabel(r'$\omega$')
plt.legend()
plt.tight_layout()
plt.show()
```
